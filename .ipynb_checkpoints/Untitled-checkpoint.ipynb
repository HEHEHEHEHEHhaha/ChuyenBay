## 1. Phát biểu bài toán học máy
- **Loại bài toán:** Học có giám sát (Supervised Learning) - Hồi quy (Regression).
- **Đơn vị mẫu (Sample/Instance):** Một chuyến bay nội địa cụ thể được lên lịch.
- **Đầu vào (Features):** Các thông tin về thời gian bay (`MONTH`, `DAY_OF_MONTH`, `DAY_OF_WEEK`, `CRS_DEP_TIME`, `CRS_ARR_TIME`, `CRS_ELAPSED_TIME`), hành trình (`ORIGIN`, `DEST`, `DISTANCE`), hãng hàng không (`OP_UNIQUE_CARRIER`), và dữ liệu thời tiết tại sân bay đi/đến (`TEMP_C`, `PRECIPITATION_MM`, `WIND_SPEED_10M_KMH`).
- **Đầu ra (Label/Target):** Độ trễ khi đến của chuyến bay tính bằng phút (`ARR_DELAY`).

## 2. Mục tiêu và tiêu chí thành công
- **Mục tiêu thực tế:** Dự đoán trước số phút trễ của chuyến bay để hỗ trợ hành khách lên kế hoạch và giúp các hãng hàng không/sân bay điều phối nguồn lực tốt hơn.
- **Mục tiêu học máy:** Xây dựng mô hình hồi quy dự đoán giá trị `ARR_DELAY` với sai số thấp nhất.
- **Chỉ số đánh giá chính:** RMSE (Root Mean Squared Error) và MAE (Mean Absolute Error).
- **Mức kết quả mong muốn:** Mô hình có RMSE thấp hơn mức baseline (ví dụ: dự đoán mọi chuyến bay trễ bằng giá trị trung bình của toàn bộ tập dữ liệu).

## 3. Nguồn và mô tả dữ liệu
- **Nguồn dữ liệu:**
  - *Chuyến bay:* Bureau of Transportation Statistics (BTS) từ tháng 09/2025 đến tháng 08/2026.
  - *Thời tiết:* Lấy từ API Open-Meteo thông qua script tự động.
- **Đặc trưng (Data Dictionary):**
  - `MONTH`, `DAY_OF_MONTH`, `DAY_OF_WEEK`: Thông tin lịch thời gian.
  - `OP_UNIQUE_CARRIER`: Mã hãng hàng không.
  - `ORIGIN`, `DEST`: Mã sân bay xuất phát và hạ cánh.
  - `CRS_DEP_TIME`, `CRS_ARR_TIME`: Giờ khởi hành và giờ đến theo kế hoạch.
  - `CRS_ELAPSED_TIME`, `DISTANCE`: Thời gian bay dự kiến và khoảng cách.
  - `ORIGIN_TEMP_C`, `ORIGIN_PRECIPITATION_MM`, `ORIGIN_WIND_SPEED_10M_KMH`: Nhiệt độ, lượng mưa, tốc độ gió tại sân bay xuất phát.
  - `DEST_TEMP_C`, `DEST_PRECIPITATION_MM`, `DEST_WIND_SPEED_10M_KMH`: Thời tiết tại sân bay đến.
  - `ARR_DELAY` (Nhãn): Số phút đến trễ (Âm = đến sớm, Dương = đến trễ).
