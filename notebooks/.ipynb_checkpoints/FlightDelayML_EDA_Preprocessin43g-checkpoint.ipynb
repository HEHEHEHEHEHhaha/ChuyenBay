# Phân tích dữ liệu chuyến bay

Notebook khảo sát dữ liệu chuyến bay đã xử lý, đối chiếu với dữ liệu thô,
trực quan hóa, kiểm tra chất lượng và chuẩn bị dữ liệu cho bài toán hồi quy
dự đoán số phút trễ đến nơi (`ARR_DELAY`).

## 1. Đọc dữ liệu đã xử lý

Nguồn đầu vào là `data/output/LocTruongDL/`.
Notebook có 16 đặc trưng và nhãn `ARR_DELAY`.

```python
from pathlib import Path
import re
import warnings
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from IPython.display import display

ROOT = Path.cwd()
CLEAN_DIR = ROOT / "data" / "output" / "LocTruongDL"
RAW_DIR = ROOT / "data" / "raw"
FIG_DIR = ROOT / "reports" / "figures"
TABLE_DIR = ROOT / "reports" / "tables"

for folder in [FIG_DIR, TABLE_DIR]:
    folder.mkdir(parents=True, exist_ok=True)

FEATURES = [
    "MONTH", "DAY_OF_MONTH", "DAY_OF_WEEK",
    "OP_UNIQUE_CARRIER", "ORIGIN", "DEST",
    "CRS_DEP_TIME", "CRS_ARR_TIME",
    "CRS_ELAPSED_TIME", "DISTANCE",
    "ORIGIN_TEMP_C", "ORIGIN_PRECIPITATION_MM",
    "ORIGIN_WIND_SPEED_10M_KMH",
    "DEST_TEMP_C", "DEST_PRECIPITATION_MM",
    "DEST_WIND_SPEED_10M_KMH"
]
TARGET = "ARR_DELAY"
EXPECTED = FEATURES + [TARGET]
CATEGORICAL = ["OP_UNIQUE_CARRIER", "ORIGIN", "DEST"]


def month_from_name(path):
    m = re.search(r"(20\d{2})[_-]T?(\d{2})", path.stem, re.I)
    if m and 1 <= int(m.group(2)) <= 12:
        return f"{int(m.group(1)):04d}-{int(m.group(2)):02d}"
    return None


def read_header(path):
    return [
        str(c).strip().upper()
        for c in pd.read_csv(
            path, nrows=0, encoding="utf-8-sig"
        ).columns
    ]


def discover_flight_csv(folder):
    if not folder.exists():
        return []

    found = []
    for p in folder.rglob("*.csv"):
        try:
            header = set(read_header(p))
            if {"ORIGIN", "DEST", "ARR_DELAY"}.issubset(header):
                found.append(p)
        except Exception as exc:
            print(f"Bỏ qua header lỗi: {p.name} ({exc})")
    return found


files = discover_flight_csv(CLEAN_DIR)
if not files:
    raise FileNotFoundError(
        f"Không tìm thấy CSV chuyến bay tại {CLEAN_DIR}. "
        "Kiểm tra thư mục data/output/LocTruongDL."
    )

files = sorted(
    files,
    key=lambda p: (month_from_name(p) or "9999-99", p.name)
)

frames = []
period_codes = []
period_labels = []
file_records = []

for path in files:
    header = read_header(path)
    missing = [c for c in EXPECTED if c not in header]
    if missing:
        raise ValueError(f"{path.name} thiếu cột: {missing}")

    part = pd.read_csv(
        path, usecols=EXPECTED,
        low_memory=False, encoding="utf-8-sig"
    )
    part.columns = [str(c).strip().upper() for c in part.columns]

    for c in FEATURES + [TARGET]:
        if c not in CATEGORICAL:
            part[c] = pd.to_numeric(part[c], errors="coerce")

    for c in CATEGORICAL:
        part[c] = (
            part[c].astype("string")
            .str.strip()
            .replace({"": pd.NA, "nan": pd.NA, "None": pd.NA})
        )

    label = month_from_name(path)
    if label is None:
        label = f"unknown-{path.stem}"

    period_labels.append(label)
    period_codes.extend([len(period_labels) - 1] * len(part))
    frames.append(part)

    file_records.append({
        "File": path.name,
        "Tháng suy ra từ tên file": label,
        "Số dòng": len(part)
    })

df = pd.concat(frames, ignore_index=True)

# Mã kỳ thời gian được lưu riêng, không thêm cột kỹ thuật vào df.
period_codes = np.asarray(period_codes, dtype=np.int16)
order = sorted(range(len(period_labels)), key=lambda i: period_labels[i])
rank = {old: new for new, old in enumerate(order)}
period_codes = np.asarray(
    [rank[i] for i in period_codes], dtype=np.int16
)
period_labels = [period_labels[i] for i in order]

file_summary = pd.DataFrame(file_records)
display(file_summary)
file_summary.to_csv(
    TABLE_DIR / "00_danh_sach_file.csv",
    index=False, encoding="utf-8-sig"
)

print(f"Tổng số mẫu sau xử lý: {len(df):,}")
print(f"Số đặc trưng đầu vào: {len(FEATURES)}")
print(f"Nhãn hồi quy: {TARGET}")
print(f"Tổng số trường bài toán: {len(EXPECTED)}")
print(f"Số file đã đọc: {len(files)}")

if len(files) != 12:
    warnings.warn(f"Tìm thấy {len(files)} file, không phải 12 file.")

expected_months = [
    "2025-09", "2025-10", "2025-11", "2025-12",
    "2026-01", "2026-02", "2026-03", "2026-04",
    "2026-05", "2026-06", "2026-07", "2026-08"
]
observed = [
    m for m in period_labels
    if re.fullmatch(r"20\d{2}-\d{2}", m)
]
if observed and observed != expected_months:
    print("Các tháng đọc được:", ", ".join(observed))

print(f"Kích thước: {df.shape[0]:,} dòng × {df.shape[1]} cột")
```

## 2. Thống kê mô tả

Bảng dưới được tính trực tiếp từ 12 file sau xử lý.
Tỷ lệ thiếu và số giá trị khác nhau giúp nhận diện các vấn đề cần khảo sát
trước khi xây dựng mô hình.

```python
profile = pd.DataFrame({
    "Kiểu dữ liệu": df.dtypes.astype(str),
    "Số thiếu": df.isna().sum(),
    "Tỷ lệ thiếu (%)": (df.isna().mean() * 100).round(3),
    "Số giá trị khác nhau": df.nunique(dropna=True)
}).reset_index(names="Đặc trưng")

display(profile)
profile.to_csv(
    TABLE_DIR / "01_mo_ta_truong_du_lieu.csv",
    index=False, encoding="utf-8-sig"
)

numeric_cols = [
    c for c in EXPECTED if c not in CATEGORICAL
]
desc = df[numeric_cols].describe(
    percentiles=[.01, .25, .5, .75, .99]
).T

display(desc.round(3))
desc.to_csv(
    TABLE_DIR / "02_thong_ke_mo_ta_so.csv",
    encoding="utf-8-sig"
)

print(
    "Đọc các phân vị và độ lệch chuẩn để đánh giá độ phân tán. "
    "Không kết luận dữ liệu sai chỉ vì giá trị nằm ở đuôi phân bố."
)
```

## 3. So sánh dữ liệu trước và sau xử lý

Đọc dữ liệu thô theo từng khối để giảm sử dụng bộ nhớ.
Phần thống kê này chỉ đáng tin khi dữ liệu thô và dữ liệu đã xử lý
bao phủ cùng phạm vi tháng.

Dữ liệu thô chưa mapping thời tiết, vì vậy không so sánh trực tiếp
giá trị thiếu thời tiết giữa hai giai đoạn như thể hai bảng có cùng điều kiện.

```python
raw_files = discover_flight_csv(RAW_DIR)
raw_records = []

critical_raw = [
    "OP_UNIQUE_CARRIER", "ORIGIN", "DEST",
    "CRS_DEP_TIME", "CRS_ARR_TIME",
    "CRS_ELAPSED_TIME", "DISTANCE", "ARR_DELAY"
]

if not raw_files:
    print("Không tìm thấy CSV chuyến bay trong data/raw/.")
    print(
        "Chưa thể xác định số dòng trước xử lý. "
        "Hãy đặt CSV thô đã giải nén vào data/raw/ rồi chạy lại cell này."
    )
else:
    for raw_path in raw_files:
        header = read_header(raw_path)
        use = [
            c for c in ["YEAR", "MONTH", "CANCELLED", "DIVERTED"]
            + critical_raw if c in header
        ]

        total = cancelled = diverted = 0
        missing_target = missing_core = eligible = 0

        try:
            for chunk in pd.read_csv(
                raw_path, usecols=use, chunksize=250_000,
                low_memory=False, encoding="utf-8-sig"
            ):
                chunk.columns = [
                    str(c).strip().upper() for c in chunk.columns
                ]
                total += len(chunk)

                if "CANCELLED" in chunk:
                    cancelled_mask = (
                        pd.to_numeric(
                            chunk["CANCELLED"], errors="coerce"
                        ).fillna(0).eq(1)
                    )
                    cancelled += int(cancelled_mask.sum())
                else:
                    cancelled_mask = pd.Series(
                        False, index=chunk.index
                    )

                if "DIVERTED" in chunk:
                    diverted_mask = (
                        pd.to_numeric(
                            chunk["DIVERTED"], errors="coerce"
                        ).fillna(0).eq(1)
                    )
                    diverted += int(diverted_mask.sum())
                else:
                    diverted_mask = pd.Series(
                        False, index=chunk.index
                    )

                if "ARR_DELAY" in chunk:
                    y_raw = pd.to_numeric(
                        chunk["ARR_DELAY"], errors="coerce"
                    )
                    missing_target += int(y_raw.isna().sum())
                else:
                    y_raw = pd.Series(
                        np.nan, index=chunk.index
                    )

                check_cols = [
                    c for c in critical_raw if c in chunk.columns
                ]
                if check_cols:
                    core_missing = chunk[check_cols].isna().any(axis=1)
                else:
                    core_missing = pd.Series(
                        False, index=chunk.index
                    )

                missing_core += int(core_missing.sum())
                eligible += int((
                    ~cancelled_mask
                    & ~diverted_mask
                    & ~core_missing
                    & y_raw.notna()
                ).sum())

            raw_records.append({
                "File thô": raw_path.name,
                "Số dòng thô": total,
                "Số chuyến hủy": cancelled,
                "Số chuyến chuyển hướng": diverted,
                "Thiếu ARR_DELAY": missing_target,
                "Thiếu ít nhất một trường cốt lõi": missing_core,
                "Đủ điều kiện theo trường thô hiện có": eligible
            })

        except Exception as exc:
            print(f"Không thống kê được {raw_path.name}: {exc}")

if raw_records:
    raw_summary = pd.DataFrame(raw_records)
    display(raw_summary)
    raw_summary.to_csv(
        TABLE_DIR / "03_thong_ke_du_lieu_tho.csv",
        index=False, encoding="utf-8-sig"
    )

    raw_total = int(raw_summary["Số dòng thô"].sum())
    clean_total = len(df)

    compare = pd.DataFrame({
        "Giai đoạn": [
            "Dữ liệu thô",
            "Dữ liệu sau xử lý"
        ],
        "Số dòng": [raw_total, clean_total]
    })
    compare["Chênh lệch so với dữ liệu thô"] = [
        0, clean_total - raw_total
    ]

    display(compare)
    compare.to_csv(
        TABLE_DIR / "04_so_sanh_truoc_sau.csv",
        index=False, encoding="utf-8-sig"
    )

    fig, ax = plt.subplots(figsize=(7, 4))
    ax.bar(compare["Giai đoạn"], compare["Số dòng"])
    ax.set_title("Quy mô dữ liệu trước và sau xử lý")
    ax.set_ylabel("Số bản ghi")
    ax.ticklabel_format(axis="y", style="plain")
    ax.tick_params(axis="x", rotation=10)
    fig.tight_layout()
    fig.savefig(
        FIG_DIR / "01_quy_mo_truoc_sau.png",
        dpi=160, bbox_inches="tight"
    )
    plt.show()
    plt.close(fig)
```

## 4. Chất lượng dữ liệu

Định lượng dữ liệu thiếu, trùng lặp, giờ không hợp lệ, sai miền giá trị
và ngoại lai. IQR chỉ dùng để phát hiện; không tự động xóa các độ trễ lớn,
vì đó có thể là những sự kiện chuyến bay thực tế.

Vì `ARR_DELAY` là nhãn liên tục của bài toán hồi quy, phần phân bố theo
khoảng độ trễ chỉ giúp hiểu dữ liệu, không thay thế vấn đề mất cân bằng lớp
của bài toán phân loại.

```python
duplicate_count = int(df[EXPECTED].duplicated().sum())
quality_rows = []

for col in EXPECTED:
    quality_rows.append({
        "Vấn đề": f"Thiếu dữ liệu: {col}",
        "Số lượng": int(df[col].isna().sum()),
        "Tỷ lệ (%)": round(float(df[col].isna().mean() * 100), 4)
    })

quality_rows.append({
    "Vấn đề": "Bản ghi trùng trên 17 trường",
    "Số lượng": duplicate_count,
    "Tỷ lệ (%)": round(duplicate_count / max(1, len(df)) * 100, 4)
})


def valid_hhmm(series):
    x = pd.to_numeric(series, errors="coerce")
    hh, mm = np.floor(x / 100), x % 100
    normal = (
        (hh >= 0) & (hh <= 23)
        & (mm >= 0) & (mm <= 59)
    )
    return x.notna() & (normal | (x == 2400))


for c in ["CRS_DEP_TIME", "CRS_ARR_TIME"]:
    invalid = int(
        (df[c].notna() & ~valid_hhmm(df[c])).sum()
    )
    quality_rows.append({
        "Vấn đề": f"Giờ HHMM không hợp lệ: {c}",
        "Số lượng": invalid,
        "Tỷ lệ (%)": round(invalid / max(1, len(df)) * 100, 4)
    })

checks = [
    ("MONTH", ~df["MONTH"].between(1, 12)),
    ("DAY_OF_MONTH", ~df["DAY_OF_MONTH"].between(1, 31)),
    ("DAY_OF_WEEK", ~df["DAY_OF_WEEK"].between(1, 7)),
    ("DISTANCE", df["DISTANCE"] < 0),
    ("CRS_ELAPSED_TIME", df["CRS_ELAPSED_TIME"] <= 0)
]

for c, condition in checks:
    bad = int((df[c].notna() & condition).sum())
    quality_rows.append({
        "Vấn đề": f"Sai miền giá trị: {c}",
        "Số lượng": bad,
        "Tỷ lệ (%)": round(bad / max(1, len(df)) * 100, 4)
    })

outlier_cols = [
    "ARR_DELAY", "DISTANCE", "CRS_ELAPSED_TIME",
    "ORIGIN_TEMP_C", "ORIGIN_PRECIPITATION_MM",
    "ORIGIN_WIND_SPEED_10M_KMH", "DEST_TEMP_C",
    "DEST_PRECIPITATION_MM", "DEST_WIND_SPEED_10M_KMH"
]

for c in outlier_cols:
    s = pd.to_numeric(df[c], errors="coerce").dropna()
    if len(s):
        q1, q3 = s.quantile([.25, .75])
        iqr = q3 - q1
        n_out = int(
            ((s < q1 - 1.5 * iqr) | (s > q3 + 1.5 * iqr)).sum()
        )
        quality_rows.append({
            "Vấn đề": f"Ngoại lai IQR cần xem xét: {c}",
            "Số lượng": n_out,
            "Tỷ lệ (%)": round(n_out / max(1, len(s)) * 100, 4)
        })

quality = pd.DataFrame(quality_rows)
display(quality)
quality.to_csv(
    TABLE_DIR / "05_kiem_tra_chat_luong.csv",
    index=False, encoding="utf-8-sig"
)

valid_y = pd.to_numeric(df[TARGET], errors="coerce").dropna()
delay_bins = pd.cut(
    valid_y,
    bins=[-np.inf, -15, 15, np.inf],
    labels=[
        "Sớm hơn trên 15 phút",
        "Trong khoảng -15 đến 15 phút",
        "Trễ trên 15 phút"
    ]
)
label_dist = (
    delay_bins.value_counts()
    .rename_axis("Khoảng ARR_DELAY")
    .reset_index(name="Số chuyến")
)
label_dist["Tỷ lệ (%)"] = (
    label_dist["Số chuyến"]
    / label_dist["Số chuyến"].sum() * 100
).round(2)

display(label_dist)
label_dist.to_csv(
    TABLE_DIR / "06_phan_bo_khoang_do_tre.csv",
    index=False, encoding="utf-8-sig"
)
```

## 5. Phân bố nhãn và đặc trưng chính

Các biểu đồ được tạo từ CSV thực tế, lưu thành ảnh PNG.
Trục độ trễ được giới hạn ở phân vị 1%–99% để nhìn rõ phần chính
của phân bố; giá trị gốc không bị xóa khỏi dữ liệu.

```python
# Chuẩn bị tập dữ liệu con để vẽ biểu đồ cho nhanh
plot_df = (
    df.sample(n=min(200_000, len(df)), random_state=42)
    if len(df) > 200_000 else df
)

# ==========================================
# 5.1. Phân bố ARR_DELAY
# ==========================================
s = pd.to_numeric(df[TARGET], errors="coerce").dropna()
lo, hi = s.quantile([.01, .99])

fig, ax = plt.subplots(figsize=(8, 4.5))
ax.hist(s.clip(lo, hi), bins=70, edgecolor="white")
ax.set(
    title="Phân bố độ trễ đến nơi (ARR_DELAY)",
    xlabel="Độ trễ (phút; hiển thị phân vị 1%–99%)",
    ylabel="Số chuyến"
)
fig.tight_layout()
fig.savefig(FIG_DIR / "02_phan_bo_arr_delay.png", dpi=160)
plt.show()
plt.close(fig)

print(
    f"Nhận xét: trung vị ARR_DELAY = {s.median():.2f} phút; "
    f"trung bình = {s.mean():.2f} phút; "
    f"P90 = {s.quantile(.90):.2f} phút. "
    "Chênh lệch trung bình và trung vị có thể cho thấy đuôi trễ lớn."
)


def hhmm_minutes(series):
    x = pd.to_numeric(series, errors="coerce")
    hh, mm = np.floor(x / 100), x % 100
    valid = (
        (((hh >= 0) & (hh <= 23) & (mm >= 0) & (mm <= 59))
         | (x == 2400))
        & x.notna()
    )
    out = pd.Series(np.nan, index=series.index, dtype=float)
    out.loc[valid] = (
        (hh[valid] % 24) * 60 + mm[valid]
    ).astype(float)
    return out


# ==========================================
# 5.2. Giờ khởi hành
# ==========================================
dep_minutes = hhmm_minutes(plot_df["CRS_DEP_TIME"])
fig, ax = plt.subplots(figsize=(8, 4.3))
ax.hist(
    dep_minutes.dropna() / 60,
    bins=np.arange(0, 25, 1),
    edgecolor="white"
)
ax.set(
    title="Phân bố giờ khởi hành theo lịch",
    xlabel="Giờ trong ngày",
    ylabel="Số chuyến",
    xticks=range(0, 25, 2)
)
fig.tight_layout()
fig.savefig(FIG_DIR / "03_phan_bo_gio_khoi_hanh.png", dpi=160)
plt.show()
plt.close(fig)

print(
    "Nhận xét: những cột cao cho biết các khung giờ có nhiều chuyến. "
    "Giờ có tính chu kỳ 24 giờ, vì vậy nên dùng sin/cos khi chuẩn bị mô hình."
)


# ==========================================
# 5.3. Khoảng cách chuyến bay
# ==========================================
dist = pd.to_numeric(plot_df["DISTANCE"], errors="coerce").dropna()
fig, ax = plt.subplots(figsize=(8, 4.3))
ax.hist(dist, bins=60, edgecolor="white")
ax.set(
    title="Phân bố khoảng cách chuyến bay",
    xlabel="Khoảng cách (dặm theo BTS)",
    ylabel="Số chuyến"
)
fig.tight_layout()
fig.savefig(FIG_DIR / "04_phan_bo_khoang_cach.png", dpi=160)
plt.show()
plt.close(fig)

print(
    f"Nhận xét: khoảng cách trung vị là {dist.median():.1f} dặm. "
    "Không nên tự động xóa các chuyến xa chỉ vì nằm ở đuôi phân bố."
)


# ==========================================
# 5.5. Phân bố số chuyến bay theo thứ trong tuần
# ==========================================
if "DAY_OF_WEEK" in plot_df.columns:
    # 1 = Thứ 2, 2 = Thứ 3, ..., 7 = Chủ nhật
    dow_counts = plot_df["DAY_OF_WEEK"].value_counts().sort_index()
    dow_labels = ["Thứ 2", "Thứ 3", "Thứ 4", "Thứ 5", "Thứ 6", "Thứ 7", "CN"]
    
    fig, ax = plt.subplots(figsize=(8, 4.3))
    # Sử dụng width=1.0 và edgecolor='white' để tạo các cột sát nhau giống biểu đồ histogram
    ax.bar(dow_labels, dow_counts.values, width=1.0, edgecolor="white")
    ax.set(
        title="Phân bố số chuyến bay theo thứ trong tuần",
        xlabel="Thứ trong tuần",
        ylabel="Số chuyến"
    )
    
    fig.tight_layout()
    fig.savefig(FIG_DIR / "06_phan_bo_chuyen_bay_theo_thu.png", dpi=160)
    plt.show()
    plt.close(fig)

    print(
        "Nhận xét: Cột cao hơn cho thấy ngày đó trong tuần có lịch trình bay dày đặc hơn. "
        "Ngày cuối tuần hoặc giữa tuần thường có lưu lượng khác biệt rõ rệt."
    )
else:
    print("Không tìm thấy cột DAY_OF_WEEK để vẽ phân bố.")
# ==========================================
# 5.4. Độ trễ theo tháng
# ==========================================
month_delay = pd.DataFrame({
    "Tháng dữ liệu": [period_labels[i] for i in period_codes],
    "ARR_DELAY": df[TARGET].to_numpy()
})
month_delay = month_delay[
    month_delay["Tháng dữ liệu"].str.match(r"20\d{2}-\d{2}", na=False)
]

if not month_delay.empty:
    monthly = (
        month_delay.groupby("Tháng dữ liệu")["ARR_DELAY"]
        .agg(So_chuyen="count", Trung_vi="median", Trung_binh="mean")
        .reset_index()
    )

    fig, ax = plt.subplots(figsize=(10, 4.5))
    ax.plot(monthly["Tháng dữ liệu"], monthly["Trung_vi"],
            marker="o", label="Trung vị")
    ax.plot(monthly["Tháng dữ liệu"], monthly["Trung_binh"],
            marker="o", label="Trung bình")
    ax.set(
        title="Độ trễ đến nơi theo tháng",
        xlabel="Tháng dữ liệu",
        ylabel="ARR_DELAY (phút)"
    )
    ax.tick_params(axis="x", rotation=45)
    ax.legend()
    fig.tight_layout()
    fig.savefig(FIG_DIR / "05_do_tre_theo_thang.png", dpi=160)
    plt.show()
    plt.close(fig)

    display(monthly)
    monthly.to_csv(
        TABLE_DIR / "07_do_tre_theo_thang.csv",
        index=False, encoding="utf-8-sig"
    )

    worst = monthly.loc[monthly["Trung_vi"].idxmax()]
    best = monthly.loc[monthly["Trung_vi"].idxmin()]
    print(
        f"Nhận xét: trung vị cao nhất ở {worst['Tháng dữ liệu']} "
        f"({worst['Trung_vi']:.2f} phút), thấp nhất ở "
        f"{best['Tháng dữ liệu']} ({best['Trung_vi']:.2f} phút). "
        "Đây là quan sát theo tháng, không tự chứng minh nguyên nhân."
    )
else:
    print("Không xác định được tháng từ tên file để vẽ xu hướng.")

```

## 6. Quan hệ giữa đặc trưng và nhãn

Dùng tương quan Spearman cho quan hệ đơn điệu giữa các biến số,
so sánh trung vị theo hãng bay và phân tích nhóm lượng mưa tại sân bay đi.
Các kết quả này là quan sát thống kê, không khẳng định quan hệ nhân quả.

```python
# 6.1. Tương quan Spearman giữa các biến số
corr_cols = [
    "ARR_DELAY", "MONTH", "DAY_OF_MONTH", "DAY_OF_WEEK",
    "CRS_DEP_TIME", "CRS_ARR_TIME",  # Đã thêm giờ cất/hạ cánh
    "CRS_ELAPSED_TIME", "DISTANCE",
    "ORIGIN_TEMP_C", "ORIGIN_PRECIPITATION_MM",
    "ORIGIN_WIND_SPEED_10M_KMH",
    "DEST_TEMP_C", "DEST_PRECIPITATION_MM",
    "DEST_WIND_SPEED_10M_KMH"
]

# Chỉ sử dụng những cột thực sự tồn tại
corr_cols = [c for c in corr_cols if c in plot_df.columns]
corr_data = plot_df[corr_cols].apply(pd.to_numeric, errors="coerce")
corr = corr_data.corr(method="spearman")

fig, ax = plt.subplots(figsize=(12, 9))
im = ax.imshow(corr, cmap="coolwarm", vmin=-1, vmax=1, aspect="auto")

ax.set_xticks(
    range(len(corr.columns)), corr.columns,
    rotation=75, ha="right"
)
ax.set_yticks(range(len(corr.index)), corr.index)
ax.set_title("Tương quan Spearman giữa các biến số")

# Hiện hệ số trên từng ô để dễ đọc
for i in range(len(corr.index)):
    for j in range(len(corr.columns)):
        value = corr.iloc[i, j]
        if pd.notna(value):
            ax.text(
                j, i, f"{value:.2f}",
                ha="center", va="center", fontsize=7
            )

fig.colorbar(im, ax=ax, label="Hệ số Spearman")
fig.tight_layout()
fig.savefig(
    FIG_DIR / "06_tuong_quan_spearman.png",
    dpi=160, bbox_inches="tight"
)
plt.show()
plt.close(fig)

corr.to_csv(
    TABLE_DIR / "08_tuong_quan_spearman.csv",
    encoding="utf-8-sig"
)

target_corr = (
    corr["ARR_DELAY"].drop("ARR_DELAY").dropna()
    .sort_values(key=lambda x: x.abs(), ascending=False)
)

print("Nhận xét 1 — Các quan hệ đơn điệu nổi bật với ARR_DELAY:")
display(target_corr.head(5).rename(
    "Spearman với ARR_DELAY"
).to_frame())

print(
    "Tương quan không chứng minh nguyên nhân; hệ số gần 0 cũng "
    "không loại trừ quan hệ phi tuyến."
)


# Hàm vẽ các biểu đồ theo nhóm và độ trễ
def plot_weather_delay(
    data, weather_col, target_col, title,
    xlabel, filename, bins, labels
):
    weather = data[[weather_col, target_col]].copy()

    weather[weather_col] = pd.to_numeric(
        weather[weather_col], errors="coerce"
    )
    weather[target_col] = pd.to_numeric(
        weather[target_col], errors="coerce"
    )
    weather = weather.dropna()

    if weather.empty:
        print(f"Không đủ dữ liệu để vẽ: {title}")
        return

    weather["Nhóm"] = pd.cut(
        weather[weather_col],
        bins=bins,
        labels=labels,
        include_lowest=True
    )

    stats = (
        weather.groupby("Nhóm", observed=False)[target_col]
        .agg(
            So_chuyen="count",
            Trung_vi="median",
            Trung_binh="mean"
        )
        .reset_index()
    )

    display(stats)

    fig, ax = plt.subplots(figsize=(9, 4.5))
    ax.bar(
        stats["Nhóm"].astype(str),
        stats["Trung_vi"]
    )

    ax.set(
        title=title,
        xlabel=xlabel,
        ylabel="Trung vị ARR_DELAY (phút)"
    )
    ax.tick_params(axis="x", rotation=20)
    fig.tight_layout()
    fig.savefig(
        FIG_DIR / filename,
        dpi=160, bbox_inches="tight"
    )
    plt.show()
    plt.close(fig)

    table_name = filename.replace(".png", ".csv")
    stats.to_csv(
        TABLE_DIR / table_name,
        index=False, encoding="utf-8-sig"
    )

    valid = stats[
        (stats["So_chuyen"] > 0) & stats["Trung_vi"].notna()
    ]

    if len(valid) >= 2:
        low = valid.loc[valid["Trung_vi"].idxmin()]
        high = valid.loc[valid["Trung_vi"].idxmax()]
        print(
            f"Nhận xét: trung vị độ trễ thấp nhất ở nhóm "
            f"{low['Nhóm']} ({low['Trung_vi']:.2f} phút), "
            f"cao nhất ở nhóm {high['Nhóm']} "
            f"({high['Trung_vi']:.2f} phút)."
        )

    print(
        "Lưu ý: đây là so sánh mô tả; kết quả còn phụ thuộc "
        "sân bay, thời điểm, mùa và các yếu tố khác."
    )


# 6.2. Độ trễ theo hãng bay; chỉ xét hãng có ít nhất 500 chuyến
carrier = df[["OP_UNIQUE_CARRIER", TARGET]].dropna()

carrier_stats = (
    carrier.groupby("OP_UNIQUE_CARRIER")[TARGET]
    .agg(So_chuyen="count", Trung_vi="median", Trung_binh="mean")
)
carrier_stats = carrier_stats.query(
    "So_chuyen >= 500"
).sort_values("Trung_vi")

if len(carrier_stats):
    show = pd.concat([
        carrier_stats.head(8), carrier_stats.tail(8)
    ]).sort_values("Trung_vi")

    fig, ax = plt.subplots(figsize=(9, 5))
    ax.barh(show.index.astype(str), show["Trung_vi"])
    ax.set(
        title="Trung vị độ trễ theo hãng bay (nhóm ≥ 500 chuyến)",
        xlabel="Trung vị ARR_DELAY (phút)",
        ylabel="Hãng bay"
    )
    fig.tight_layout()
    fig.savefig(
        FIG_DIR / "07_do_tre_theo_hang.png",
        dpi=160, bbox_inches="tight"
    )
    plt.show()
    plt.close(fig)

    display(
        carrier_stats.sort_values(
            "Trung_vi", ascending=False
        ).head(10)
    )

    hi = carrier_stats["Trung_vi"].idxmax()
    low = carrier_stats["Trung_vi"].idxmin()

    print(
        f"Nhận xét 2 — {hi} có trung vị trễ cao nhất "
        f"({carrier_stats.loc[hi, 'Trung_vi']:.2f} phút); "
        f"{low} thấp nhất "
        f"({carrier_stats.loc[low, 'Trung_vi']:.2f} phút)."
    )
else:
    print("Không có hãng nào đủ 500 chuyến để so sánh.")


# 6.3. Nhóm lượng mưa tại sân bay khởi hành
plot_weather_delay(
    df,
    "ORIGIN_PRECIPITATION_MM",
    TARGET,
    "Độ trễ theo lượng mưa tại sân bay khởi hành",
    "Lượng mưa (mm)",
    "08_mua_va_do_tre.png",
    [-0.001, 0, 1, 5, np.inf],
    ["Không mưa", ">0–1 mm", ">1–5 mm", ">5 mm"]
)


# 6.4. Nhiệt độ tại sân bay khởi hành
plot_weather_delay(
    df,
    "ORIGIN_TEMP_C",
    TARGET,
    "Độ trễ theo nhiệt độ tại sân bay khởi hành",
    "Nhiệt độ (°C)",
    "10_nhiet_do_san_bay_di.png",
    [-np.inf, 0, 10, 20, 30, np.inf],
    ["<0°C", "0–10°C", "10–20°C", "20–30°C", ">30°C"]
)


# 6.5. Tốc độ gió tại sân bay khởi hành
plot_weather_delay(
    df,
    "ORIGIN_WIND_SPEED_10M_KMH",
    TARGET,
    "Độ trễ theo tốc độ gió tại sân bay khởi hành",
    "Tốc độ gió ở độ cao 10 m (km/h)",
    "11_gio_san_bay_di.png",
    [-0.001, 10, 20, 30, 40, np.inf],
    ["0–10", ">10–20", ">20–30", ">30–40", ">40 km/h"]
)


# 6.6. Lượng mưa tại sân bay đến
plot_weather_delay(
    df,
    "DEST_PRECIPITATION_MM",
    TARGET,
    "Độ trễ theo lượng mưa tại sân bay đến",
    "Lượng mưa (mm)",
    "12_mua_san_bay_den.png",
    [-0.001, 0, 1, 5, np.inf],
    ["Không mưa", ">0–1 mm", ">1–5 mm", ">5 mm"]
)


# 6.7. Nhiệt độ tại sân bay đến
plot_weather_delay(
    df,
    "DEST_TEMP_C",
    TARGET,
    "Độ trễ theo nhiệt độ tại sân bay đến",
    "Nhiệt độ (°C)",
    "13_nhiet_do_san_bay_den.png",
    [-np.inf, 0, 10, 20, 30, np.inf],
    ["<0°C", "0–10°C", "10–20°C", "20–30°C", ">30°C"]
)


# 6.8. Tốc độ gió tại sân bay đến
plot_weather_delay(
    df,
    "DEST_WIND_SPEED_10M_KMH",
    TARGET,
    "Độ trễ theo tốc độ gió tại sân bay đến",
    "Tốc độ gió ở độ cao 10 m (km/h)",
    "14_gio_san_bay_den.png",
    [-0.001, 10, 20, 30, 40, np.inf],
    ["0–10", ">10–20", ">20–30", ">30–40", ">40 km/h"]
)

# 6.9. Độ trễ theo khung giờ cất cánh (CRS_DEP_TIME)
plot_weather_delay(
    df,
    "CRS_DEP_TIME",
    TARGET,
    "Độ trễ theo khung giờ cất cánh theo lịch",
    "Khung giờ cất cánh (Định dạng HHMM)",
    "15_do_tre_theo_gio_cat_canh.png",
    [-1, 600, 1200, 1800, 2400],
    ["Đêm (00h-06h)", "Sáng (06h-12h)", "Chiều (12h-18h)", "Tối (18h-24h)"]
)

# 6.10. Độ trễ theo khung giờ hạ cánh (CRS_ARR_TIME)
plot_weather_delay(
    df,
    "CRS_ARR_TIME",
    TARGET,
    "Độ trễ theo khung giờ hạ cánh theo lịch",
    "Khung giờ hạ cánh (Định dạng HHMM)",
    "16_do_tre_theo_gio_ha_canh.png",
    [-1, 600, 1200, 1800, 2400],
    ["Đêm (00h-06h)", "Sáng (06h-12h)", "Chiều (12h-18h)", "Tối (18h-24h)"]
)
```

## 7. Chuẩn bị dữ liệu cho mô hình hồi quy

Nhãn là `ARR_DELAY`. Tỷ lệ phân chia: Tách ngẫu nhiên theo tỷ lệ 70% Train, 15% Validation, và 15% Test. Sau đó lưu vào data/ouput có 3 thư mục trong đây là train, validation, test.

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from pathlib import Path

# 1. Định nghĩa đường dẫn
INPUT_DIR = ROOT / "data" / "output" / "LocTruongDL"
OUTPUT_DIR = ROOT / "data" / "output"
TRAIN_DIR = OUTPUT_DIR / "train"
VAL_DIR = OUTPUT_DIR / "val"
TEST_DIR = OUTPUT_DIR / "test"

# Tạo các thư mục đầu ra nếu chưa có
for folder in [TRAIN_DIR, VAL_DIR, TEST_DIR]:
    folder.mkdir(parents=True, exist_ok=True)

# 2. Duyệt qua từng file CSV trong thư mục LocTruongDL
csv_files = list(INPUT_DIR.glob("*.csv"))
print(f"Tìm thấy {len(csv_files)} file. Đang tiến hành chia dữ liệu...\n")

for file_path in csv_files:
    print(f"Đang xử lý: {file_path.name}")
    
    # Đọc dữ liệu của tháng đó
    df_month = pd.read_csv(file_path)
    
    # Bỏ qua nếu file trống
    if df_month.empty:
        continue
        
    # Chia 70% Train, 30% Temp
    df_train, df_temp = train_test_split(df_month, test_size=0.3, random_state=42)
    # Chia Temp thành 15% Val, 15% Test
    df_val, df_test = train_test_split(df_temp, test_size=0.5, random_state=42)
    
    # 3. Lưu các tập vào thư mục tương ứng, giữ nguyên tên file
    df_train.to_csv(TRAIN_DIR / file_path.name, index=False, encoding="utf-8-sig")
    df_val.to_csv(VAL_DIR / file_path.name, index=False, encoding="utf-8-sig")
    df_test.to_csv(TEST_DIR / file_path.name, index=False, encoding="utf-8-sig")

print("\nHoàn tất chia và lưu dữ liệu cho tất cả các tháng!")
```
