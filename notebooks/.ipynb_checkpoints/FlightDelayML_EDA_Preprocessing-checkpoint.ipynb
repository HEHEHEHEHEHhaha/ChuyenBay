**Mục tiêu dự đoán:** `ARR_DELAY` — số phút trễ khi đến nơi. Giá trị âm có nghĩa là đến sớm; giá trị dương có nghĩa là đến trễ. Không chuyển nhãn thành phân loại trong mô hình hồi quy.

## 1. Khởi tạo môi trường và đường dẫn

Chạy Notebook từ thư mục gốc `FlightDelayML`. Mặc định Notebook tìm dữ liệu đã xử lý tại `data/output/LocTruongDL/` (đúng theo đường dẫn dự án đã dùng trước đó), và tìm dữ liệu thô trong `data/raw/`. Nếu thư mục hiện tại khác, hãy sửa `PROJECT_ROOT` ở cell dưới.

```python
from pathlib import Path
import re
import sys
import math
import warnings
import importlib.util
from collections import Counter

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from IPython.display import display, Markdown

SEED = 42
CHUNK_SIZE = 200_000
PLOT_SAMPLE_SIZE = 200_000

# Mặc định: Notebook nằm ở thư mục gốc FlightDelayML.
PROJECT_ROOT = Path.cwd().resolve()
# Nếu Jupyter khởi chạy từ thư mục con, dò ngược lên thư mục có data/.
for candidate in [PROJECT_ROOT, *PROJECT_ROOT.parents]:
    if (candidate / "data").exists():
        PROJECT_ROOT = candidate
        break

RAW_DIR = PROJECT_ROOT / "data" / "raw"
PROCESSED_DIR_CANDIDATES = [
    PROJECT_ROOT / "data" / "output" / "LocTruongDL",
    PROJECT_ROOT / "data" / "processed",
    PROJECT_ROOT / "data" / "output" / "processed",
]
REPORTS_DIR = PROJECT_ROOT / "reports"
FIGURES_DIR = REPORTS_DIR / "figures"
TABLES_DIR = REPORTS_DIR / "tables"
FIGURES_DIR.mkdir(parents=True, exist_ok=True)
TABLES_DIR.mkdir(parents=True, exist_ok=True)

print("Python:", sys.version.split()[0])
print("Pandas:", pd.__version__)
print("Thư mục dự án:", PROJECT_ROOT)
print("Dữ liệu thô:", RAW_DIR)
print("Thư mục hình xuất ra:", FIGURES_DIR)
```

## 2. Khai báo đặc trưng, nhãn và các hàm đọc dữ liệu

Các đặc trưng dự kiến theo thiết kế của nhóm gồm 16 cột: ngày/thứ, hãng bay, sân bay đi/đến, giờ khởi hành/đến theo lịch, thời lượng bay theo lịch, khoảng cách và 6 đặc trưng thời tiết. `ARR_DELAY` là nhãn hồi quy.

Notebook chỉ nạp các cột cần cho phân tích và chuẩn bị mô hình để giảm RAM; từ điển dữ liệu vẫn được lấy từ toàn bộ tiêu đề cột của file đã xử lý.

```python
TARGET = "ARR_DELAY"
MODEL_FEATURES = [
    "MONTH", "DAY_OF_MONTH", "DAY_OF_WEEK",
    "OP_UNIQUE_CARRIER", "ORIGIN", "DEST",
    "CRS_DEP_TIME", "CRS_ARR_TIME", "CRS_ELAPSED_TIME", "DISTANCE",
    "ORIGIN_TEMP_C", "ORIGIN_PRECIPITATION_MM", "ORIGIN_WIND_SPEED_10M_KMH",
    "DEST_TEMP_C", "DEST_PRECIPITATION_MM", "DEST_WIND_SPEED_10M_KMH",
]
CATEGORICAL_FEATURES = ["OP_UNIQUE_CARRIER", "ORIGIN", "DEST"]
NUMERIC_FEATURES = [c for c in MODEL_FEATURES if c not in CATEGORICAL_FEATURES]
WEATHER_FEATURES = [c for c in MODEL_FEATURES if c.startswith(("ORIGIN_", "DEST_"))]

AUDIT_COLUMNS = [
    "FL_DATE", "YEAR", "QUARTER", "MONTH", "DAY_OF_MONTH", "DAY_OF_WEEK",
    "OP_UNIQUE_CARRIER", "OP_CARRIER", "OP_CARRIER_AIRLINE_ID", "TAIL_NUM",
    "OP_CARRIER_FL_NUM", "ORIGIN", "DEST", "CRS_DEP_TIME", "DEP_TIME",
    "DEP_DELAY", "ARR_TIME", "ARR_DELAY", "ARR_DEL15", "CANCELLED",
    "CANCELLATION_CODE", "DIVERTED", "CRS_ARR_TIME", "CRS_ELAPSED_TIME",
    "ACTUAL_ELAPSED_TIME", "AIR_TIME", "DISTANCE",
]
LOAD_COLUMNS = sorted(set(MODEL_FEATURES + AUDIT_COLUMNS + [
    "DEP_DELAY_NEW", "DEP_DEL15", "ARR_DELAY_NEW", "ARR_DELAY_GROUP",
    "DEP_DELAY_GROUP", "WHEELS_OFF", "WHEELS_ON", "TAXI_OUT", "TAXI_IN",
    "ORIGIN_TEMP_C", "ORIGIN_RELATIVE_HUMIDITY_PCT", "ORIGIN_DEW_POINT_C",
    "ORIGIN_APPARENT_TEMP_C", "ORIGIN_PRECIPITATION_MM", "ORIGIN_RAIN_MM",
    "ORIGIN_SNOWFALL_CM", "ORIGIN_WEATHER_CODE", "ORIGIN_PRESSURE_MSL_HPA",
    "ORIGIN_CLOUD_COVER_PCT", "ORIGIN_WIND_SPEED_10M_KMH",
    "ORIGIN_WIND_DIRECTION_10M_DEG", "ORIGIN_WIND_GUSTS_10M_KMH",
    "DEST_TEMP_C", "DEST_RELATIVE_HUMIDITY_PCT", "DEST_DEW_POINT_C",
    "DEST_APPARENT_TEMP_C", "DEST_PRECIPITATION_MM", "DEST_RAIN_MM",
    "DEST_SNOWFALL_CM", "DEST_WEATHER_CODE", "DEST_PRESSURE_MSL_HPA",
    "DEST_CLOUD_COVER_PCT", "DEST_WIND_SPEED_10M_KMH",
    "DEST_WIND_DIRECTION_10M_DEG", "DEST_WIND_GUSTS_10M_KMH",
]) )
SUPPORTED_SUFFIXES = (".csv", ".csv.gz", ".parquet", ".pq", ".xlsx", ".xls")


def normalise_column_name(name):
    return str(name).replace("\ufeff", "").strip().upper()


def get_suffix_kind(path):
    name = path.name.lower()
    if name.endswith(".csv.gz"):
        return "csv"
    if name.endswith(".csv"):
        return "csv"
    if name.endswith((".parquet", ".pq")):
        return "parquet"
    if name.endswith((".xlsx", ".xls")):
        return "excel"
    return None


def list_supported_files(directory):
    if not directory.exists():
        return []
    result = []
    for path in directory.rglob("*"):
        if not path.is_file() or path.name.startswith("~$") or path.name.startswith("."):
            continue
        if get_suffix_kind(path):
            result.append(path)
    return sorted(result)


def read_header(path):
    kind = get_suffix_kind(path)
    if kind == "csv":
        return list(pd.read_csv(path, nrows=0, encoding_errors="replace").columns)
    if kind == "parquet":
        # Parquet giữ được schema mà không cần nạp toàn bộ bản ghi.
        try:
            import pyarrow.parquet as pq
            return pq.ParquetFile(path).schema.names
        except Exception:
            return list(pd.read_parquet(path).columns)
    if kind == "excel":
        return list(pd.read_excel(path, nrows=0).columns)
    return []


def read_selected(path, wanted=None, nrows=None):
    """Đọc cột chọn lọc và chuẩn hóa tên cột về UPPERCASE."""
    kind = get_suffix_kind(path)
    original_header = read_header(path)
    norm_to_original = {}
    for col in original_header:
        norm_to_original.setdefault(normalise_column_name(col), col)
    if wanted is None:
        chosen = original_header
    else:
        chosen = [norm_to_original[c] for c in wanted if c in norm_to_original]
    if kind == "csv":
        df = pd.read_csv(path, usecols=chosen, low_memory=False, encoding_errors="replace", nrows=nrows)
    elif kind == "parquet":
        df = pd.read_parquet(path, columns=chosen)
        if nrows is not None:
            df = df.head(nrows)
    elif kind == "excel":
        df = pd.read_excel(path, usecols=chosen, nrows=nrows)
    else:
        raise ValueError(f"Định dạng chưa hỗ trợ: {path}")
    df.columns = [normalise_column_name(c) for c in df.columns]
    # Một số bản dữ liệu BTS có OP_CARRIER thay cho OP_UNIQUE_CARRIER.
    if "OP_UNIQUE_CARRIER" not in df.columns and "OP_CARRIER" in df.columns:
        df["OP_UNIQUE_CARRIER"] = df["OP_CARRIER"]
    return df


def month_key_from_name(filename):
    stem = Path(filename).stem
    # Tên kiểu 2025_T09.csv, 2025_9.csv, ...
    m = re.search(r"(20\d{2})[^0-9]{0,8}(0?[1-9]|1[0-2])(?!\d)", stem, flags=re.I)
    if m:
        return f"{int(m.group(1)):04d}-{int(m.group(2)):02d}"
    month_names = {
        "JAN": 1, "FEB": 2, "MAR": 3, "APR": 4, "MAY": 5, "JUN": 6,
        "JUL": 7, "AUG": 8, "SEP": 9, "OCT": 10, "NOV": 11, "DEC": 12,
    }
    y = re.search(r"20\d{2}", stem)
    mon = re.search(r"(?i)(JAN|FEB|MAR|APR|MAY|JUN|JUL|AUG|SEP|OCT|NOV|DEC)", stem)
    if y and mon:
        return f"{int(y.group(0)):04d}-{month_names[mon.group(1).upper()]:02d}"
    return None


def is_flight_file(path):
    try:
        cols = {normalise_column_name(c) for c in read_header(path)}
        return {"ARR_DELAY", "ORIGIN", "DEST"}.issubset(cols)
    except Exception as exc:
        print(f"Bỏ qua file không đọc được tiêu đề {path.name}: {exc}")
        return False

print("Số đặc trưng dự kiến:", len(MODEL_FEATURES))
print("Nhãn:", TARGET)
```

## 3. Tìm 12 file đã xử lý và file chuyến bay thô

Notebook dò file có các cột `ARR_DELAY`, `ORIGIN`, `DEST`; vì vậy file thời tiết độc lập sẽ không bị nhầm thành file chuyến bay. Kiểm tra danh sách bên dưới trước khi chạy tiếp. Nếu thư mục dữ liệu của mày khác cấu hình, hãy sửa `PROCESSED_DIR_CANDIDATES` hoặc `RAW_DIR` ở phần 1.

```python
processed_files = []
chosen_processed_dir = None
for folder in PROCESSED_DIR_CANDIDATES:
    candidates = [p for p in list_supported_files(folder) if is_flight_file(p)]
    if candidates:
        processed_files = candidates
        chosen_processed_dir = folder
        break

# Nếu chưa tìm thấy ở các thư mục thường dùng, dò dưới data/ nhưng không dò vào data/raw.
if not processed_files:
    data_dir = PROJECT_ROOT / "data"
    if data_dir.exists():
        candidates = [
            p for p in list_supported_files(data_dir)
            if RAW_DIR not in p.parents and is_flight_file(p)
        ]
        if candidates:
            processed_files = candidates
            chosen_processed_dir = candidates[0].parent

raw_files = [p for p in list_supported_files(RAW_DIR) if is_flight_file(p)]

processed_file_inventory = []
for p in processed_files:
    try:
        cols = read_header(p)
        processed_file_inventory.append({
            "File": str(p.relative_to(PROJECT_ROOT)),
            "Tháng theo tên file": month_key_from_name(p.name) or "Không suy ra được",
            "Số cột trong file": len(cols),
            "Có ARR_DELAY": "ARR_DELAY" in {normalise_column_name(c) for c in cols},
            "Dung lượng (MB)": round(p.stat().st_size / (1024**2), 2),
        })
    except Exception as exc:
        print("Không đọc được:", p, exc)

print(f"Thư mục file đã xử lý được chọn: {chosen_processed_dir}")
print(f"Tìm thấy {len(processed_files)} file chuyến bay đã xử lý.")
print(f"Tìm thấy {len(raw_files)} file chuyến bay thô trong {RAW_DIR}.")
display(pd.DataFrame(processed_file_inventory))

if len(processed_files) == 0:
    raise FileNotFoundError(
        "Không tìm thấy file chuyến bay đã xử lý. Hãy đặt CSV vào data/output/LocTruongDL/ "
        "hoặc sửa PROCESSED_DIR_CANDIDATES ở cell đầu. File cần có ARR_DELAY, ORIGIN, DEST."
    )
if len(processed_files) != 12:
    print(f"\nCẢNH BÁO: tìm thấy {len(processed_files)} file, không phải 12. Kiểm tra có thiếu/thừa file hay không.")
```

## 4. Tạo bộ dữ liệu phân tích từ 12 file sau xử lý

Cell này nối dữ liệu đã xử lý để khảo sát tổng thể. Nó chỉ nạp cột cần dùng; file đầu vào gốc trên đĩa không bị sửa. Các dòng thiếu nhãn được giữ để **báo cáo chất lượng**, sau đó mới loại khỏi tập chuẩn bị mô hình nếu cần.

```python
processed_parts = []
all_processed_headers = {}

for i, path in enumerate(processed_files, start=1):
    print(f"[{i}/{len(processed_files)}] Đọc {path.name} ...")
    all_processed_headers[path.name] = [normalise_column_name(c) for c in read_header(path)]
    part = read_selected(path, wanted=LOAD_COLUMNS)
    part["_SOURCE_FILE"] = path.name
    part["_FILE_MONTH_KEY"] = month_key_from_name(path.name) or ""
    processed_parts.append(part)

df_after_original_types = pd.concat(processed_parts, ignore_index=True, sort=False)
del processed_parts

# Đảm bảo biến alias hãng bay đã được chuẩn hóa trong cả dữ liệu đã nối.
if "OP_UNIQUE_CARRIER" not in df_after_original_types.columns and "OP_CARRIER" in df_after_original_types.columns:
    df_after_original_types["OP_UNIQUE_CARRIER"] = df_after_original_types["OP_CARRIER"]

df_after = df_after_original_types
print(f"Tổng số dòng sau xử lý đang khảo sát: {len(df_after):,}")
print(f"Số cột được nạp cho phân tích: {len(df_after.columns):,}")
print("Các cột đầu:", df_after.columns[:20].tolist())
```

## 5. Nguồn, điều kiện sử dụng và mô tả dữ liệu

- **Dữ liệu chuyến bay:** nhóm tải từ hệ thống dữ liệu On-Time Performance của U.S. Bureau of Transportation Statistics (BTS), nếu đúng với nguồn mà nhóm đã sử dụng. Trang nguồn: [BTS TranStats — On-Time : Reporting Carrier On-Time Performance](https://transtats.bts.gov/Fields.asp?gnoyr_VQ=FGJ).
- **Dữ liệu thời tiết:** dữ liệu lịch sử theo thời gian và tọa độ được lấy từ [Open-Meteo Historical Weather API](https://open-meteo.com/en/docs/historical-weather-api), theo script tải thời tiết của dự án.
- **Điều kiện sử dụng:** trước khi nộp, cần kiểm tra điều khoản, giấy phép và yêu cầu ghi nhận nguồn hiện hành của từng nguồn; không tự khẳng định dữ liệu được phép tái phân phối nếu chưa kiểm tra. ZIP nộp cho thầy nên kèm URL nguồn và ngày tải/phiên bản dữ liệu nếu đã ghi nhận.
- **Phạm vi dữ liệu dự kiến:** 09/2025–08/2026. Notebook sẽ thống kê số mẫu và số cột thực tế, không điền số liệu ước lượng vào báo cáo.

Nếu nguồn thực tế khác phần mô tả trên, hãy sửa đoạn Markdown này cho trùng với nguồn mà nhóm đã tải.

```python
# Tổng quan thực tế sau xử lý
source_summary = pd.DataFrame([{
    "Số file đã xử lý": len(processed_files),
    "Tổng số mẫu (dòng)": len(df_after),
    "Số đặc trưng mô hình theo thiết kế": len(MODEL_FEATURES),
    "Nhãn hồi quy": TARGET,
    "Ngày bắt đầu (nếu đọc được)": None,
    "Ngày kết thúc (nếu đọc được)": None,
}])

flight_date = pd.Series(pd.NaT, index=df_after.index, dtype="datetime64[ns]")
if "FL_DATE" in df_after.columns:
    flight_date = pd.to_datetime(df_after["FL_DATE"], errors="coerce")
if flight_date.notna().sum() == 0 and {"YEAR", "MONTH", "DAY_OF_MONTH"}.issubset(df_after.columns):
    flight_date = pd.to_datetime({
        "year": pd.to_numeric(df_after["YEAR"], errors="coerce"),
        "month": pd.to_numeric(df_after["MONTH"], errors="coerce"),
        "day": pd.to_numeric(df_after["DAY_OF_MONTH"], errors="coerce"),
    }, errors="coerce")

df_after["_FLIGHT_DATE"] = flight_date
source_summary.loc[0, "Ngày bắt đầu (nếu đọc được)"] = flight_date.min()
source_summary.loc[0, "Ngày kết thúc (nếu đọc được)"] = flight_date.max()
display(source_summary)
source_summary.to_csv(TABLES_DIR / "01_tong_quan_du_lieu.csv", index=False, encoding="utf-8-sig")

present_model_features = [c for c in MODEL_FEATURES if c in df_after.columns]
missing_model_features = [c for c in MODEL_FEATURES if c not in df_after.columns]
print("\nĐặc trưng có trong dữ liệu:", present_model_features)
if missing_model_features:
    print("THIẾU các đặc trưng so với thiết kế:", missing_model_features)
    print("Cần kiểm tra lại file đã xử lý hoặc cập nhật danh sách đặc trưng; không nên im lặng bỏ qua các cột này.")
if TARGET not in df_after.columns:
    raise KeyError(f"Không tìm thấy cột nhãn {TARGET} trong dữ liệu đã xử lý.")
```

## 6. So sánh trước xử lý và sau xử lý — phần bắt buộc

Phần này đọc từng file thô theo từng khối (chunk) để không cần nạp toàn bộ 12 tháng vào RAM cùng lúc. Nó thống kê các vấn đề có thể đo trực tiếp từ dữ liệu gốc: số dòng, số dòng hủy/chuyển hướng, thiếu nhãn, thiếu ít nhất một đặc trưng chuyến bay có sẵn trong file thô và khóa chuyến bay trùng.

Sau đó đối chiếu với file đã xử lý cùng tháng: số dòng còn lại, thiếu nhãn/đặc trưng cuối cùng, dòng trùng theo khóa chuyến bay. **Số dòng chênh lệch là mức giảm ròng**, không được cộng các lý do loại dòng một cách máy móc vì một chuyến bay có thể đồng thời vi phạm nhiều điều kiện.

```python
RAW_CORE_COLUMNS = [
    "MONTH", "DAY_OF_MONTH", "DAY_OF_WEEK", "OP_UNIQUE_CARRIER", "OP_CARRIER",
    "ORIGIN", "DEST", "CRS_DEP_TIME", "CRS_ARR_TIME", "CRS_ELAPSED_TIME", "DISTANCE",
]
KEY_CANDIDATES = ["FL_DATE", "OP_UNIQUE_CARRIER", "OP_CARRIER_FL_NUM", "ORIGIN", "DEST", "CRS_DEP_TIME"]


def as_numeric(series):
    return pd.to_numeric(series, errors="coerce")


def count_duplicate_keys_in_frame(frame):
    # Khóa tự nhiên gần đúng của một chuyến bay theo ngày, hãng, số hiệu, tuyến và giờ theo lịch.
    keys = [c for c in KEY_CANDIDATES if c in frame.columns]
    if len(keys) >= 4 and "ORIGIN" in keys and "DEST" in keys:
        # Không tính khóa bị thiếu như một bản ghi xác định duy nhất.
        valid = frame[keys].notna().all(axis=1)
        return int(frame.loc[valid].duplicated(subset=keys).sum()), keys
    return int(frame.duplicated().sum()), ["toàn bộ cột đang có (dự phòng)"]


def scan_raw_file(path):
    """Thống kê file raw theo chunk; khóa duplicate dùng các cột định danh nếu có."""
    wanted = set(LOAD_COLUMNS) | set(RAW_CORE_COLUMNS) | set(KEY_CANDIDATES) | {"CANCELLED", "DIVERTED", "ARR_DELAY"}
    if get_suffix_kind(path) != "csv":
        d = read_selected(path, wanted=wanted)
        frames = [d]
    else:
        header = read_header(path)
        available = {normalise_column_name(c) for c in header}
        use_cols = [c for c in header if normalise_column_name(c) in wanted]
        frames = pd.read_csv(path, usecols=use_cols, chunksize=CHUNK_SIZE, low_memory=False, encoding_errors="replace")

    total = cancelled = diverted = target_missing = core_missing = 0
    key_parts = []
    core_seen = []
    for chunk in frames:
        chunk.columns = [normalise_column_name(c) for c in chunk.columns]
        if "OP_UNIQUE_CARRIER" not in chunk.columns and "OP_CARRIER" in chunk.columns:
            chunk["OP_UNIQUE_CARRIER"] = chunk["OP_CARRIER"]
        n = len(chunk)
        total += n
        if "CANCELLED" in chunk.columns:
            cancelled += int(as_numeric(chunk["CANCELLED"]).eq(1).sum())
        if "DIVERTED" in chunk.columns:
            diverted += int(as_numeric(chunk["DIVERTED"]).eq(1).sum())
        if TARGET in chunk.columns:
            target_missing += int(chunk[TARGET].isna().sum())
        core_cols = [c for c in ["MONTH", "DAY_OF_MONTH", "DAY_OF_WEEK", "OP_UNIQUE_CARRIER", "ORIGIN", "DEST", "CRS_DEP_TIME", "CRS_ARR_TIME", "CRS_ELAPSED_TIME", "DISTANCE"] if c in chunk.columns]
        if core_cols:
            core_missing += int(chunk[core_cols].isna().any(axis=1).sum())
        keys = [c for c in KEY_CANDIDATES if c in chunk.columns]
        if len(keys) >= 4 and "ORIGIN" in keys and "DEST" in keys:
            # Chỉ giữ các cột khóa nhỏ; dữ liệu mỗi file được xử lý riêng để giới hạn RAM.
            key_parts.append(chunk[keys].copy())
        elif not core_seen:
            core_seen = [c for c in chunk.columns]

    duplicate_keys = np.nan
    key_method = "Không đủ khóa định danh"
    if key_parts:
        key_frame = pd.concat(key_parts, ignore_index=True)
        keys = [c for c in KEY_CANDIDATES if c in key_frame.columns]
        if len(keys) >= 4 and "ORIGIN" in keys and "DEST" in keys:
            valid = key_frame[keys].notna().all(axis=1)
            duplicate_keys = int(key_frame.loc[valid].duplicated(subset=keys).sum())
            key_method = "Khóa ngày + hãng + số hiệu + tuyến + giờ lịch (các cột có sẵn)"
        del key_frame, key_parts

    return {
        "Tháng": month_key_from_name(path.name) or f"Tên file: {path.name}",
        "File thô": path.name,
        "Số dòng trước xử lý": total,
        "Dòng hủy (CANCELLED=1)": cancelled,
        "Dòng chuyển hướng (DIVERTED=1)": diverted,
        "Thiếu nhãn ARR_DELAY": target_missing,
        "Thiếu ít nhất 1 đặc trưng chuyến bay có sẵn": core_missing,
        "Dòng trùng khóa chuyến bay": duplicate_keys,
        "Phương pháp kiểm tra trùng": key_method,
    }


raw_summary_records = []
if raw_files:
    for i, path in enumerate(raw_files, start=1):
        print(f"Quét dữ liệu thô [{i}/{len(raw_files)}]: {path.name}")
        try:
            raw_summary_records.append(scan_raw_file(path))
        except Exception as exc:
            warnings.warn(f"Không thể thống kê file raw {path.name}: {exc}")
else:
    print(
        "KHÔNG TÌM THẤY FILE CHUYẾN BAY THÔ. Không thể tính số liệu trước xử lý thật. "
        "Hãy đặt các file BTS gốc đã giải nén vào data/raw/ rồi chạy lại từ đầu."
    )

raw_summary_df = pd.DataFrame(raw_summary_records)
if not raw_summary_df.empty:
    display(raw_summary_df)
    raw_summary_df.to_csv(TABLES_DIR / "02_thong_ke_truoc_xu_ly.csv", index=False, encoding="utf-8-sig")
```

## 7. Thống kê từng file sau xử lý và bảng so sánh trước–sau

Bảng sau xử lý dùng cùng một khóa tháng từ tên file. Nếu tên file không cho biết tháng, Notebook giữ khóa theo tên file để tránh ghép nhầm. Hãy kiểm tra kết quả ghép có đúng 12 cặp tháng hay không.

```python
processed_summary_records = []
for path in processed_files:
    part = df_after.loc[df_after["_SOURCE_FILE"].eq(path.name)]
    n = len(part)
    feature_cols_present = [c for c in MODEL_FEATURES if c in part.columns]
    missing_feature_rows = int(part[feature_cols_present].isna().any(axis=1).sum()) if feature_cols_present else np.nan
    duplicate_count, duplicate_method = count_duplicate_keys_in_frame(part)
    processed_summary_records.append({
        "Tháng": month_key_from_name(path.name) or f"Tên file: {path.name}",
        "File sau xử lý": path.name,
        "Số dòng sau xử lý": n,
        "Thiếu nhãn ARR_DELAY sau xử lý": int(part[TARGET].isna().sum()) if TARGET in part else np.nan,
        "Thiếu ít nhất 1 đặc trưng mô hình sau xử lý": missing_feature_rows,
        "Dòng trùng khóa chuyến bay sau xử lý": duplicate_count,
        "Phương pháp kiểm tra trùng sau xử lý": ", ".join(duplicate_method),
        "Số đặc trưng dự kiến còn thiếu trong file": len([c for c in MODEL_FEATURES if c not in part.columns]),
    })
processed_summary_df = pd.DataFrame(processed_summary_records)
display(processed_summary_df)
processed_summary_df.to_csv(TABLES_DIR / "03_thong_ke_sau_xu_ly.csv", index=False, encoding="utf-8-sig")

if not raw_summary_df.empty:
    before_after_df = raw_summary_df.merge(
        processed_summary_df, on="Tháng", how="outer", validate="one_to_one"
    )
    before_after_df["Chênh lệch số dòng (trước - sau)"] = (
        before_after_df["Số dòng trước xử lý"] - before_after_df["Số dòng sau xử lý"]
    )
    before_after_df["Tỷ lệ giảm ròng (%)"] = (
        before_after_df["Chênh lệch số dòng (trước - sau)"]
        / before_after_df["Số dòng trước xử lý"].replace(0, np.nan) * 100
    ).round(2)
    before_after_df["Tỷ lệ giữ lại (%)"] = (
        before_after_df["Số dòng sau xử lý"]
        / before_after_df["Số dòng trước xử lý"].replace(0, np.nan) * 100
    ).round(2)
    cols_first = [
        "Tháng", "File thô", "File sau xử lý", "Số dòng trước xử lý", "Số dòng sau xử lý",
        "Chênh lệch số dòng (trước - sau)", "Tỷ lệ giảm ròng (%)", "Tỷ lệ giữ lại (%)",
        "Dòng hủy (CANCELLED=1)", "Dòng chuyển hướng (DIVERTED=1)",
        "Thiếu nhãn ARR_DELAY", "Thiếu ít nhất 1 đặc trưng chuyến bay có sẵn",
        "Dòng trùng khóa chuyến bay", "Thiếu nhãn ARR_DELAY sau xử lý",
        "Thiếu ít nhất 1 đặc trưng mô hình sau xử lý", "Dòng trùng khóa chuyến bay sau xử lý",
    ]
    display(before_after_df[[c for c in cols_first if c in before_after_df.columns]].sort_values("Tháng"))
    before_after_df.to_csv(TABLES_DIR / "04_so_sanh_truoc_sau.csv", index=False, encoding="utf-8-sig")

    matched = before_after_df["Số dòng trước xử lý"].notna() & before_after_df["Số dòng sau xử lý"].notna()
    print(f"\nSố cặp tháng ghép được: {int(matched.sum())}/{len(before_after_df)}")
    if matched.sum() != 12:
        print("CẢNH BÁO: chưa ghép đủ 12 cặp tháng. Kiểm tra tên file, tháng dữ liệu và vị trí thư mục.")
else:
    before_after_df = pd.DataFrame()
    print("Chưa tạo được bảng so sánh trước–sau vì thiếu dữ liệu thô. Bảng sau xử lý vẫn được xuất.")
```

### Nhận xét tự động về mức giảm số dòng

Nhận xét dưới đây được sinh từ số liệu vừa tính. Nó mô tả chênh lệch tổng thể, **không thay thế việc kiểm tra từng nguyên nhân loại dòng trong script tiền xử lý**. Nếu số dòng sau xử lý nhiều hơn trước, cần kiểm tra xem có ghép trùng khi mapping thời tiết hay không.

```python
if not before_after_df.empty:
    valid_pairs = before_after_df.dropna(subset=["Số dòng trước xử lý", "Số dòng sau xử lý"])
    if len(valid_pairs):
        total_raw = valid_pairs["Số dòng trước xử lý"].sum()
        total_after = valid_pairs["Số dòng sau xử lý"].sum()
        net_removed = total_raw - total_after
        pct_removed = (net_removed / total_raw * 100) if total_raw else np.nan
        print(f"Tổng số dòng ở các file raw ghép được: {total_raw:,.0f}")
        print(f"Tổng số dòng ở file đã xử lý ghép được: {total_after:,.0f}")
        print(f"Chênh lệch ròng: {net_removed:,.0f} dòng ({pct_removed:.2f}% so với raw).")
        if net_removed > 0:
            print("Diễn giải: số dòng đầu ra giảm. Đối chiếu các cột hủy, chuyển hướng và thiếu dữ liệu để kiểm chứng nguyên nhân.")
        elif net_removed < 0:
            print("CẢNH BÁO: đầu ra nhiều dòng hơn raw. Cần kiểm tra ghép thời tiết có làm nhân bản một chuyến bay hoặc hai bảng có phạm vi khác nhau không.")
        else:
            print("Số dòng tổng thể không đổi; điều đó không tự chứng minh dữ liệu không được làm sạch. Kiểm tra từng vấn đề chất lượng và khóa trùng.")
else:
    print("Không có dữ liệu raw để sinh nhận xét so sánh số dòng trước–sau.")
```

## 9. Thống kê mô tả và phân bố các đặc trưng chính

Trước khi vẽ biểu đồ, chuyển các trường được dùng như số sang kiểu số; giá trị không chuyển được sẽ thành thiếu và được thống kê ở phần kiểm tra chất lượng. Không xóa các ngoại lai chỉ vì chúng nằm ngoài IQR: một chuyến bay trễ rất lâu có thể là quan sát có thật và có ý nghĩa.

```python
NUMERIC_AUDIT_COLUMNS = sorted(set(NUMERIC_FEATURES + [
    "ARR_DELAY", "DEP_DELAY", "CRS_ELAPSED_TIME", "DISTANCE", "CANCELLED", "DIVERTED",
    "YEAR", "MONTH", "DAY_OF_MONTH", "DAY_OF_WEEK", "CRS_DEP_TIME", "CRS_ARR_TIME",
]))
conversion_issue_rows = []
for col in NUMERIC_AUDIT_COLUMNS:
    if col not in df_after.columns:
        continue
    original = df_after[col]
    converted = pd.to_numeric(original, errors="coerce")
    # Phân biệt giá trị vốn thiếu với giá trị có nội dung nhưng không chuyển được sang số.
    invalid = original.notna() & converted.isna()
    conversion_issue_rows.append({
        "Cột": col,
        "Giá trị không chuyển được sang số": int(invalid.sum()),
        "Tỷ lệ trên toàn bộ dòng (%)": round(float(invalid.mean() * 100), 4) if len(invalid) else 0.0,
    })
    df_after[col] = converted.astype("float32")
conversion_issues_df = pd.DataFrame(conversion_issue_rows)
display(conversion_issues_df)
conversion_issues_df.to_csv(TABLES_DIR / "06_loi_chuyen_doi_kieu_so.csv", index=False, encoding="utf-8-sig")

present_numeric = [c for c in NUMERIC_FEATURES + [TARGET] if c in df_after.columns]
descriptive_stats = df_after[present_numeric].describe(percentiles=[.01, .25, .5, .75, .99]).T
missing_counts = df_after[present_numeric].isna().sum().rename("Số thiếu")
descriptive_stats = descriptive_stats.join(missing_counts)
display(descriptive_stats.round(3))
descriptive_stats.to_csv(TABLES_DIR / "07_thong_ke_mo_ta.csv", encoding="utf-8-sig")

plot_df = df_after.sample(n=min(PLOT_SAMPLE_SIZE, len(df_after)), random_state=SEED) if len(df_after) else df_after.copy()
```

### 9.1. Phân bố nhãn `ARR_DELAY`

Biểu đồ này giúp nhìn độ lệch của nhãn, sự xuất hiện của các chuyến đến sớm (giá trị âm) và đuôi phải của các chuyến trễ lớn. Do bài toán là hồi quy, nhãn chính vẫn là số phút trễ liên tục.

```python
if TARGET in plot_df.columns and plot_df[TARGET].notna().any():
    fig, ax = plt.subplots(figsize=(10, 5))
    values = plot_df[TARGET].dropna()
    ax.hist(values, bins=100, edgecolor="white", linewidth=0.2)
    ax.set_title("Phân bố số phút trễ khi đến nơi (ARR_DELAY)")
    ax.set_xlabel("Số phút so với giờ đến theo lịch (phút)")
    ax.set_ylabel("Số chuyến bay trong mẫu vẽ")
    ax.grid(axis="y", alpha=0.25)
    fig.tight_layout()
    fig.savefig(FIGURES_DIR / "01_phan_bo_arr_delay.png", dpi=160, bbox_inches="tight")
    plt.show()
    print(
        f"Nhận xét: trung vị ARR_DELAY = {values.median():.2f} phút; "
        f"25%–75% nằm trong [{values.quantile(.25):.2f}, {values.quantile(.75):.2f}] phút; "
        f"tỷ lệ chuyến có ARR_DELAY >= 15 phút = {(values.ge(15).mean()*100):.2f}% trong mẫu vẽ. "
        "Giá trị âm là đến sớm, vì vậy không nên cắt các giá trị âm về 0 trong bài toán hồi quy."
    )
else:
    print("Không có giá trị ARR_DELAY hợp lệ để vẽ.")
```

### 9.2. Phân bố các đặc trưng số

Các biểu đồ được sinh cho các đặc trưng có trong dữ liệu. Nhận xét tự động ghi trung vị, khoảng tứ phân vị và tỷ lệ thiếu — cần được diễn giải cùng nghiệp vụ, không tự động coi các giá trị cực trị là sai.

```python
features_for_hist = [c for c in NUMERIC_FEATURES if c in plot_df.columns]
for col in features_for_hist:
    vals = plot_df[col].dropna()
    if vals.empty:
        print(f"Bỏ qua {col}: không có giá trị hợp lệ.")
        continue
    fig, ax = plt.subplots(figsize=(8, 4))
    ax.hist(vals, bins=50, edgecolor="white", linewidth=0.2)
    ax.set_title(f"Phân bố {col}")
    ax.set_xlabel(col)
    ax.set_ylabel("Số chuyến bay trong mẫu vẽ")
    ax.grid(axis="y", alpha=0.25)
    fig.tight_layout()
    fig.savefig(FIGURES_DIR / f"phan_bo_{col.lower()}.png", dpi=150, bbox_inches="tight")
    plt.show()
    missing_pct = df_after[col].isna().mean() * 100
    print(
        f"Nhận xét {col}: n mẫu hợp lệ vẽ = {len(vals):,}; trung vị = {vals.median():.3f}; "
        f"Q1 = {vals.quantile(.25):.3f}; Q3 = {vals.quantile(.75):.3f}; "
        f"thiếu trong toàn bộ dữ liệu sau xử lý = {missing_pct:.3f}%. "
        "Hãy xem xét đơn vị và miền giá trị trước khi quyết định xử lý ngoại lai."
    )
```

### 9.3. Phân bố hãng bay, ngày trong tuần và tháng

Biểu đồ đếm cho thấy dữ liệu có tập trung vào một số hãng hoặc một số thời điểm hay không. Số lượng mẫu khác nhau giữa nhóm có thể ảnh hưởng đến độ ổn định của trung bình/median.

```python
if "OP_UNIQUE_CARRIER" in df_after.columns:
    carrier_counts = df_after["OP_UNIQUE_CARRIER"].astype("string").value_counts(dropna=False).head(20)
    fig, ax = plt.subplots(figsize=(10, 5))
    carrier_counts.sort_values().plot(kind="barh", ax=ax)
    ax.set_title("20 hãng bay có số bản ghi nhiều nhất")
    ax.set_xlabel("Số chuyến bay")
    ax.set_ylabel("Mã hãng")
    fig.tight_layout()
    fig.savefig(FIGURES_DIR / "phan_bo_hang_bay.png", dpi=160, bbox_inches="tight")
    plt.show()
    print("Nhận xét: phân bố hãng bay không đồng đều nếu một số hãng chiếm phần lớn bản ghi; cần xem số mẫu từng nhóm trước khi so sánh độ trễ.")

for col, title, file_name in [
    ("DAY_OF_WEEK", "Số chuyến theo thứ trong tuần", "phan_bo_thu_trong_tuan.png"),
    ("MONTH", "Số chuyến theo tháng", "phan_bo_thang.png"),
]:
    if col in df_after.columns:
        counts = df_after[col].value_counts(dropna=False).sort_index()
        fig, ax = plt.subplots(figsize=(8, 4))
        counts.plot(kind="bar", ax=ax)
        ax.set_title(title)
        ax.set_xlabel(col)
        ax.set_ylabel("Số chuyến bay")
        fig.tight_layout()
        fig.savefig(FIGURES_DIR / file_name, dpi=160, bbox_inches="tight")
        plt.show()
        print(f"Nhận xét: biểu đồ {col} giúp kiểm tra mức độ bao phủ theo thời gian; nếu có tháng/thứ có rất ít mẫu, kết luận của nhóm đó kém ổn định hơn.")
```

## 10. Phân tích quan hệ giữa đặc trưng và nhãn

Phần này đưa ra số liệu cụ thể từ tập đang khảo sát, không chỉ in các nhận xét chung. Tương quan chỉ phản ánh quan hệ thống kê, không khẳng định quan hệ nhân quả.

```python
# Spearman trên mẫu để giảm thời gian/RAM; phù hợp hơn Pearson khi biến có phân bố lệch hoặc quan hệ đơn điệu.
relationship_cols = [c for c in NUMERIC_FEATURES + [TARGET] if c in plot_df.columns]
rel_sample = plot_df[relationship_cols].copy()
for c in relationship_cols:
    rel_sample[c] = pd.to_numeric(rel_sample[c], errors="coerce")
correlation = rel_sample.corr(method="spearman", numeric_only=True)
if TARGET in correlation.columns:
    target_corr = correlation[TARGET].drop(TARGET).dropna().sort_values(key=lambda x: x.abs(), ascending=False)
    display(target_corr.rename("Spearman rho với ARR_DELAY").to_frame())
    target_corr.rename("Spearman rho với ARR_DELAY").to_csv(TABLES_DIR / "08_tuong_quan_spearman_voi_nhan.csv", encoding="utf-8-sig")

    fig, ax = plt.subplots(figsize=(11, 8))
    im = ax.imshow(correlation, aspect="auto", vmin=-1, vmax=1, cmap="coolwarm")
    ax.set_xticks(range(len(correlation.columns)), labels=correlation.columns, rotation=90)
    ax.set_yticks(range(len(correlation.index)), labels=correlation.index)
    ax.set_title("Ma trận tương quan Spearman trên mẫu khảo sát")
    fig.colorbar(im, ax=ax, label="Spearman rho")
    fig.tight_layout()
    fig.savefig(FIGURES_DIR / "ma_tran_tuong_quan_spearman.png", dpi=160, bbox_inches="tight")
    plt.show()

    if len(target_corr):
        best = target_corr.index[0]
        print(f"Nhận xét 1: Trong các biến số được khảo sát, {best} có |Spearman rho| lớn nhất với ARR_DELAY ({target_corr.iloc[0]:.3f}). Đây là quan hệ đơn điệu trong mẫu, không chứng minh nguyên nhân và không đo hết tác động phi tuyến.")
else:
    print("Chưa đủ cột số hợp lệ để tính tương quan với nhãn.")

# So sánh median/mean/count theo hãng, tháng, thứ và giờ khởi hành dự kiến.
if {TARGET, "OP_UNIQUE_CARRIER"}.issubset(df_after.columns):
    carrier_delay = df_after.groupby("OP_UNIQUE_CARRIER", dropna=False)[TARGET].agg(["count", "median", "mean"]).query("count >= 100").sort_values("median", ascending=False)
    display(carrier_delay.head(15))
    carrier_delay.to_csv(TABLES_DIR / "09_do_tre_theo_hang_bay.csv", encoding="utf-8-sig")
    if len(carrier_delay):
        high = carrier_delay.iloc[0]
        low = carrier_delay.sort_values("median").iloc[0]
        print(f"Nhận xét 2: Trong các hãng có ít nhất 100 bản ghi, median ARR_DELAY cao nhất là {carrier_delay.index[0]} ({high['median']:.2f} phút; n={int(high['count']):,}), thấp nhất là {carrier_delay.sort_values('median').index[0]} ({low['median']:.2f} phút). Cần cân nhắc khác biệt tuyến bay và quy mô mẫu trước khi quy kết cho hãng.")

if {TARGET, "DAY_OF_WEEK"}.issubset(df_after.columns):
    weekday_delay = df_after.groupby("DAY_OF_WEEK", dropna=False)[TARGET].agg(["count", "median", "mean"]).sort_index()
    display(weekday_delay)
    weekday_delay.to_csv(TABLES_DIR / "10_do_tre_theo_thu.csv", encoding="utf-8-sig")
    if len(weekday_delay.dropna(subset=["median"])):
        wmax = weekday_delay["median"].idxmax()
        wmin = weekday_delay["median"].idxmin()
        print(f"Nhận xét 3: Median ARR_DELAY theo DAY_OF_WEEK cao nhất ở mã ngày {wmax} ({weekday_delay.loc[wmax, 'median']:.2f} phút) và thấp nhất ở mã {wmin} ({weekday_delay.loc[wmin, 'median']:.2f} phút). Cần xác nhận quy ước mã thứ trong nguồn BTS.")

# Giờ khởi hành theo lịch: chuyển HHMM thành giờ trước khi nhóm.
def hhmm_to_minutes(series):
    v = pd.to_numeric(series, errors="coerce")
    hour = np.floor(v / 100)
    minute = v % 100
    valid = v.notna() & hour.between(0, 23) & minute.between(0, 59)
    result = pd.Series(np.nan, index=series.index, dtype="float32")
    result.loc[valid] = (hour.loc[valid] * 60 + minute.loc[valid]).astype("float32")
    return result

if {TARGET, "CRS_DEP_TIME"}.issubset(df_after.columns):
    dep_minutes = hhmm_to_minutes(df_after["CRS_DEP_TIME"])
    dep_hour = (dep_minutes // 60).rename("Giờ khởi hành theo lịch")
    hour_table = df_after.assign(_DEP_HOUR=dep_hour).groupby("_DEP_HOUR", dropna=True)[TARGET].agg(["count", "median", "mean"]).sort_index()
    display(hour_table)
    hour_table.to_csv(TABLES_DIR / "11_do_tre_theo_gio_khoi_hanh.csv", encoding="utf-8-sig")
    if len(hour_table):
        max_hour = hour_table["median"].idxmax()
        print(f"Nhận xét 4: Trong các giờ có dữ liệu hợp lệ, giờ khởi hành theo lịch có median ARR_DELAY cao nhất là {int(max_hour):02d}:00 ({hour_table.loc[max_hour, 'median']:.2f} phút). Đây là tín hiệu cần kiểm tra thêm theo hãng, sân bay và mùa.")

# Tác động thời tiết: nhóm lượng mưa tại sân bay đi thành các khoảng để so sánh robust hơn.
weather_col = "ORIGIN_PRECIPITATION_MM"
if {TARGET, weather_col}.issubset(df_after.columns):
    rain_data = df_after[[TARGET, weather_col]].copy()
    rain_data[weather_col] = pd.to_numeric(rain_data[weather_col], errors="coerce")
    rain_data = rain_data.dropna()
    if len(rain_data):
        rain_data["Nhóm lượng mưa tại sân bay đi"] = pd.cut(
            rain_data[weather_col], bins=[-0.000001, 0, 1, 5, 20, np.inf],
            labels=["0 mm", ">0–1 mm", ">1–5 mm", ">5–20 mm", ">20 mm"], include_lowest=True
        )
        rain_table = rain_data.groupby("Nhóm lượng mưa tại sân bay đi", observed=False)[TARGET].agg(["count", "median", "mean"])
        display(rain_table)
        rain_table.to_csv(TABLES_DIR / "12_do_tre_theo_luong_mua_san_bay_di.csv", encoding="utf-8-sig")
        fig, ax = plt.subplots(figsize=(9, 4))
        rain_table["median"].plot(kind="bar", ax=ax)
        ax.set_title("Median ARR_DELAY theo nhóm lượng mưa tại sân bay đi")
        ax.set_xlabel("Nhóm lượng mưa")
        ax.set_ylabel("Median ARR_DELAY (phút)")
        fig.tight_layout()
        fig.savefig(FIGURES_DIR / "quan_he_mua_va_do_tre.png", dpi=160, bbox_inches="tight")
        plt.show()
        print("Nhận xét: đối chiếu median và số mẫu giữa các nhóm mưa; không kết luận nhân quả nếu chưa kiểm soát hãng, sân bay, mùa và thời điểm.")
```

## 11. Kiểm tra chất lượng dữ liệu sau xử lý

Bảng kiểm tra dưới đây định lượng các vấn đề còn lại trong 12 file đã xử lý: thiếu dữ liệu, bản ghi trùng theo khóa chuyến bay, sai kiểu số, ngày sai định dạng, ngoại lai theo IQR và mất cân bằng của nhãn phân loại phụ `ARR_DELAY >= 15`.

**Lưu ý cho hồi quy:** `ARR_DELAY` là biến liên tục nên không có “mất cân bằng lớp” theo nghĩa phân loại. Tỷ lệ trễ từ 15 phút trở lên chỉ được đưa ra như chẩn đoán phụ để hiểu tỷ lệ chuyến trễ đáng kể; nhãn mô hình chính vẫn là số phút `ARR_DELAY`.

```python
quality_rows = []
for col in [c for c in MODEL_FEATURES + [TARGET] if c in df_after.columns]:
    miss = int(df_after[col].isna().sum())
    quality_rows.append({
        "Vấn đề": "Giá trị thiếu",
        "Cột/đối tượng": col,
        "Số lượng": miss,
        "Tỷ lệ (%)": round(miss / len(df_after) * 100, 4) if len(df_after) else np.nan,
        "Diễn giải/đề xuất": "Điều tra nguyên nhân; nếu còn thiếu ở X, dùng imputer fit trên train hoặc xử lý theo nghiệp vụ; thiếu y thì không dùng hàng đó để huấn luyện có giám sát.",
    })

dup_count, dup_method = count_duplicate_keys_in_frame(df_after)
quality_rows.append({
    "Vấn đề": "Bản ghi trùng theo khóa chuyến bay",
    "Cột/đối tượng": ", ".join(dup_method),
    "Số lượng": dup_count,
    "Tỷ lệ (%)": round(dup_count / len(df_after) * 100, 4) if len(df_after) else np.nan,
    "Diễn giải/đề xuất": "Kiểm tra nguồn và mapping thời tiết; chỉ xóa sau khi xác nhận đây là cùng chuyến bay bị nhân bản, không chỉ dựa vào việc các đặc trưng giống nhau.",
})

invalid_total = int(conversion_issues_df["Giá trị không chuyển được sang số"].sum()) if len(conversion_issues_df) else 0
quality_rows.append({
    "Vấn đề": "Sai định dạng/giá trị không chuyển được sang số",
    "Cột/đối tượng": "Các cột số đã kiểm tra",
    "Số lượng": invalid_total,
    "Tỷ lệ (%)": round(invalid_total / max(1, len(df_after) * max(1, len(conversion_issues_df))) * 100, 4),
    "Diễn giải/đề xuất": "Kiểm tra giá trị nguồn và quy ước mã; không mặc định coi mọi chuỗi không chuyển được là 0.",
})

if "_FLIGHT_DATE" in df_after.columns:
    invalid_dates = int(df_after["_FLIGHT_DATE"].isna().sum())
    quality_rows.append({
        "Vấn đề": "Ngày bay thiếu/sai định dạng",
        "Cột/đối tượng": "FL_DATE hoặc YEAR/MONTH/DAY_OF_MONTH",
        "Số lượng": invalid_dates,
        "Tỷ lệ (%)": round(invalid_dates / len(df_after) * 100, 4) if len(df_after) else np.nan,
        "Diễn giải/đề xuất": "Chuẩn hóa định dạng ngày; cần ngày hợp lệ để chia tập theo thời gian.",
    })

outlier_rows = []
for col in [c for c in NUMERIC_FEATURES + [TARGET] if c in df_after.columns]:
    vals = pd.to_numeric(df_after[col], errors="coerce").dropna()
    if len(vals) < 4:
        continue
    q1, q3 = vals.quantile([.25, .75])
    iqr = q3 - q1
    lower, upper = q1 - 1.5 * iqr, q3 + 1.5 * iqr
    n_out = int(((vals < lower) | (vals > upper)).sum())
    outlier_rows.append({
        "Cột": col, "Q1": q1, "Q3": q3, "IQR": iqr,
        "Ngưỡng dưới": lower, "Ngưỡng trên": upper,
        "Số ngoại lai IQR": n_out,
        "Tỷ lệ ngoại lai (%)": round(n_out / len(vals) * 100, 4),
        "Quyết định": "Giữ để điều tra; chỉ loại nếu chứng minh là lỗi ghi nhận. Có thể dùng phép biến đổi/ước lượng robust.",
    })
outlier_df = pd.DataFrame(outlier_rows)
if len(outlier_df):
    display(outlier_df.sort_values("Tỷ lệ ngoại lai (%)", ascending=False))
    outlier_df.to_csv(TABLES_DIR / "13_ngoai_lai_iqr.csv", index=False, encoding="utf-8-sig")

if TARGET in df_after.columns:
    y_valid = pd.to_numeric(df_after[TARGET], errors="coerce").dropna()
    count_15 = int(y_valid.ge(15).sum())
    count_below = int(y_valid.lt(15).sum())
    class_balance = pd.DataFrame([
        {"Nhóm chẩn đoán phụ": "ARR_DELAY >= 15 phút", "Số dòng": count_15, "Tỷ lệ (%)": count_15 / max(1, len(y_valid)) * 100},
        {"Nhóm chẩn đoán phụ": "ARR_DELAY < 15 phút", "Số dòng": count_below, "Tỷ lệ (%)": count_below / max(1, len(y_valid)) * 100},
    ])
    display(class_balance.round(3))
    class_balance.to_csv(TABLES_DIR / "14_ty_le_nhan_15_phut_chi_de_chan_doan.csv", index=False, encoding="utf-8-sig")
    print("Đây là tỷ lệ của một ngưỡng phụ để tham khảo, không biến bài toán chính thành phân loại.")

quality_df = pd.DataFrame(quality_rows)
display(quality_df)
quality_df.to_csv(TABLES_DIR / "15_bang_van_de_chat_luong_du_lieu.csv", index=False, encoding="utf-8-sig")

fig, ax = plt.subplots(figsize=(10, 5))
missing_pct = df_after[[c for c in MODEL_FEATURES + [TARGET] if c in df_after.columns]].isna().mean().sort_values(ascending=False) * 100
missing_pct.plot(kind="bar", ax=ax)
ax.set_title("Tỷ lệ giá trị thiếu theo đặc trưng trong dữ liệu sau xử lý")
ax.set_ylabel("Tỷ lệ thiếu (%)")
ax.set_xlabel("Đặc trưng")
plt.xticks(rotation=75, ha="right")
fig.tight_layout()
fig.savefig(FIGURES_DIR / "ty_le_thieu_sau_xu_ly.png", dpi=160, bbox_inches="tight")
plt.show()
print("Nhận xét: ưu tiên kiểm tra các cột có tỷ lệ thiếu cao nhất; việc thiếu thời tiết có thể do không tìm được sân bay/giờ mapping hoặc khác quy ước múi giờ.")
```

## 12. Kiểm tra nguy cơ rò rỉ dữ liệu (data leakage)

Với mục tiêu dự đoán **trước giờ khởi hành**, chỉ nên dùng thông tin có thể biết tại thời điểm dự đoán. Một số cột có trong dữ liệu nguồn chỉ hình thành sau khi chuyến bay diễn ra, ví dụ `DEP_DELAY`, `ARR_TIME`, `WHEELS_OFF`, `WHEELS_ON`, `ACTUAL_ELAPSED_TIME`, `AIR_TIME`, các trường nguyên nhân độ trễ và trạng thái hủy/chuyển hướng.

Ngoài ra, cần kiểm tra riêng cách ghép thời tiết: nhiệt độ/mưa **quan sát thực tế tại sân bay đến vào giờ đến theo lịch** chưa chắc đã biết tại thời điểm trước giờ khởi hành. Nếu dùng dữ liệu quan sát lịch sử thay vì dự báo thời tiết có sẵn vào thời điểm đó, đây là nguy cơ rò rỉ thông tin tương lai.

```python
POST_EVENT_COLUMNS = [
    "DEP_TIME", "DEP_DELAY", "DEP_DELAY_NEW", "DEP_DEL15", "DEP_DELAY_GROUP",
    "TAXI_OUT", "WHEELS_OFF", "WHEELS_ON", "TAXI_IN", "ARR_TIME", "ARR_DELAY_NEW",
    "ARR_DEL15", "ARR_DELAY_GROUP", "CANCELLED", "CANCELLATION_CODE", "DIVERTED",
    "ACTUAL_ELAPSED_TIME", "AIR_TIME", "CARRIER_DELAY", "WEATHER_DELAY", "NAS_DELAY",
    "SECURITY_DELAY", "LATE_AIRCRAFT_DELAY", "FIRST_DEP_TIME", "TOTAL_ADD_GTIME",
    "LONGEST_ADD_GTIME", "DIV_AIRPORT_LANDINGS", "DIV_REACHED_DEST", "DIV_ACTUAL_ELAPSED_TIME",
    "DIV_ARR_DELAY", "DIV_DISTANCE",
]
leakage_rows = []
for col in POST_EVENT_COLUMNS:
    if col in df_after.columns:
        leakage_rows.append({
            "Cột hiện diện trong dữ liệu": col,
            "Nguy cơ": "Thông tin sau thời điểm dự đoán",
            "Quyết định": "Loại khỏi X khi dự đoán trước giờ khởi hành; giữ làm trường kiểm tra/nhãn phụ nếu cần.",
        })
for col in ["ORIGIN_TEMP_C", "ORIGIN_PRECIPITATION_MM", "ORIGIN_WIND_SPEED_10M_KMH", "DEST_TEMP_C", "DEST_PRECIPITATION_MM", "DEST_WIND_SPEED_10M_KMH"]:
    if col in df_after.columns:
        if col.startswith("DEST_"):
            risk = "Cần xác minh thời điểm khả dụng: thời tiết quan sát ở sân bay đến tại giờ đến lịch có thể thuộc tương lai so với lúc dự đoán. Ưu tiên dự báo thời tiết được phát hành trước giờ khởi hành."
        else:
            risk = "Cần xác minh thời điểm chốt dự báo: thời tiết quan sát gần giờ khởi hành chỉ hợp lệ nếu đã biết tại thời điểm mô hình được sử dụng."
        leakage_rows.append({"Cột hiện diện trong dữ liệu": col, "Nguy cơ": risk, "Quyết định": "Ghi rõ thời điểm dự đoán và thời điểm thời tiết; không coi dữ liệu lịch sử quan sát là đương nhiên khả dụng trong tương lai."})
leakage_df = pd.DataFrame(leakage_rows)
display(leakage_df)
leakage_df.to_csv(TABLES_DIR / "16_kiem_tra_ro_ri_du_lieu.csv", index=False, encoding="utf-8-sig")
print("Bộ X bên dưới chỉ dùng các đặc trưng đã khai báo trong MODEL_FEATURES; các trường hậu nghiệm không được đưa vào pipeline.")
```

## 13. Chuẩn bị dữ liệu cho mô hình hồi quy

### Quyết định xử lý

- Nhãn `y = ARR_DELAY`; loại các hàng thiếu nhãn khỏi tập huấn luyện vì đây là học có giám sát.
- Giữ ngày bay để chia tập theo thời gian, nhưng không đưa `FL_DATE` trực tiếp vào X.
- Các biến phân loại `OP_UNIQUE_CARRIER`, `ORIGIN`, `DEST` được mã hóa One-Hot; giá trị phân loại mới ở validation/test được xử lý bằng `handle_unknown="ignore"`.
- Các biến số được điền khuyết bằng median và chuẩn hóa. Các tham số điền khuyết/chuẩn hóa chỉ được `fit` trên train.
- `CRS_DEP_TIME` và `CRS_ARR_TIME` có định dạng HHMM, không phải số đo tuyến tính. Notebook chuyển thành sin/cos theo chu kỳ 24 giờ.
- Chia theo thời gian: train 09/2025–04/2026, validation 05/2026–06/2026, test 07/2026–08/2026. Nếu ngày thực tế không đủ các khoảng này, cần cập nhật mốc chia nhưng vẫn giữ thứ tự thời gian.

Việc dùng dữ liệu tương lai cho test phù hợp hơn mục tiêu triển khai thực tế so với chia ngẫu nhiên, vốn có thể cho kết quả lạc quan do trộn các giai đoạn thời gian.

```python
# Kiểm tra thiếu cột bắt buộc trước khi xây dựng X/y.
missing_required = [c for c in MODEL_FEATURES + [TARGET] if c not in df_after.columns]
if missing_required:
    raise KeyError(
        "Thiếu cột bắt buộc để chuẩn bị mô hình: " + ", ".join(missing_required) +
        ". Hãy kiểm tra đúng thư mục 12 file đã xử lý hoặc sửa MODEL_FEATURES."
    )

model_df = df_after.copy()
model_df[TARGET] = pd.to_numeric(model_df[TARGET], errors="coerce")
rows_before_target_filter = len(model_df)
model_df = model_df.dropna(subset=[TARGET]).copy()
rows_removed_missing_target = rows_before_target_filter - len(model_df)

# Tạo biến tuần hoàn từ giờ HHMM; thời gian lỗi sẽ thành NaN để imputer xử lý trong pipeline.
for time_col, prefix in [("CRS_DEP_TIME", "DEP"), ("CRS_ARR_TIME", "ARR")]:
    minutes = hhmm_to_minutes(model_df[time_col])
    model_df[f"{prefix}_TIME_SIN"] = np.sin(2 * np.pi * minutes / 1440).astype("float32")
    model_df[f"{prefix}_TIME_COS"] = np.cos(2 * np.pi * minutes / 1440).astype("float32")

# Mô hình dùng hai biến thời gian tuần hoàn thay cho hai số HHMM ban đầu.
MODEL_NUMERIC_FINAL = [
    "MONTH", "DAY_OF_MONTH", "DAY_OF_WEEK", "CRS_ELAPSED_TIME", "DISTANCE",
    "ORIGIN_TEMP_C", "ORIGIN_PRECIPITATION_MM", "ORIGIN_WIND_SPEED_10M_KMH",
    "DEST_TEMP_C", "DEST_PRECIPITATION_MM", "DEST_WIND_SPEED_10M_KMH",
    "DEP_TIME_SIN", "DEP_TIME_COS", "ARR_TIME_SIN", "ARR_TIME_COS",
]
MODEL_CATEGORICAL_FINAL = ["OP_UNIQUE_CARRIER", "ORIGIN", "DEST"]
MODEL_FEATURES_FINAL = MODEL_NUMERIC_FINAL + MODEL_CATEGORICAL_FINAL

# Chuyển biến phân loại về chuỗi; giá trị thiếu giữ ở dạng NaN để imputer xử lý.
for col in MODEL_CATEGORICAL_FINAL:
    model_df[col] = model_df[col].astype("object")
    model_df.loc[model_df[col].isna(), col] = np.nan

# Xác nhận cột ngày cho chia tập.
if "_FLIGHT_DATE" not in model_df.columns or model_df["_FLIGHT_DATE"].notna().sum() == 0:
    raise ValueError("Không tạo được ngày bay hợp lệ từ FL_DATE hoặc YEAR/MONTH/DAY_OF_MONTH; không thể chia tập theo thời gian.")

valid_date_count = int(model_df["_FLIGHT_DATE"].notna().sum())
invalid_date_rows = len(model_df) - valid_date_count
model_df = model_df.loc[model_df["_FLIGHT_DATE"].notna()].copy()

validation_start = pd.Timestamp("2026-05-01")
test_start = pd.Timestamp("2026-07-01")
train_mask = model_df["_FLIGHT_DATE"] < validation_start
valid_mask = (model_df["_FLIGHT_DATE"] >= validation_start) & (model_df["_FLIGHT_DATE"] < test_start)
test_mask = model_df["_FLIGHT_DATE"] >= test_start

train_df = model_df.loc[train_mask].copy()
valid_df = model_df.loc[valid_mask].copy()
test_df = model_df.loc[test_mask].copy()

split_summary = pd.DataFrame([
    {"Tập": "Train", "Bắt đầu": train_df["_FLIGHT_DATE"].min(), "Kết thúc": train_df["_FLIGHT_DATE"].max(), "Số mẫu": len(train_df), "Tỷ lệ (%)": len(train_df) / max(1, len(model_df)) * 100},
    {"Tập": "Validation", "Bắt đầu": valid_df["_FLIGHT_DATE"].min(), "Kết thúc": valid_df["_FLIGHT_DATE"].max(), "Số mẫu": len(valid_df), "Tỷ lệ (%)": len(valid_df) / max(1, len(model_df)) * 100},
    {"Tập": "Test", "Bắt đầu": test_df["_FLIGHT_DATE"].min(), "Kết thúc": test_df["_FLIGHT_DATE"].max(), "Số mẫu": len(test_df), "Tỷ lệ (%)": len(test_df) / max(1, len(model_df)) * 100},
])
display(split_summary)
split_summary.to_csv(TABLES_DIR / "17_kich_thuoc_cac_tap_theo_thoi_gian.csv", index=False, encoding="utf-8-sig")
print(f"Số dòng loại vì thiếu nhãn ARR_DELAY: {rows_removed_missing_target:,}")
print(f"Số dòng loại khỏi chia tập vì ngày bay không hợp lệ: {invalid_date_rows:,}")
print("Số đặc trưng đầu vào trước mã hóa:", len(MODEL_FEATURES_FINAL))
print("Số đặc trưng số:", len(MODEL_NUMERIC_FINAL), "| số đặc trưng phân loại:", len(MODEL_CATEGORICAL_FINAL))

if len(train_df) == 0 or len(valid_df) == 0 or len(test_df) == 0:
    print("CẢNH BÁO: Có tập rỗng. Kiểm tra phạm vi ngày thực tế và sửa validation_start/test_start; không đổi sang chia ngẫu nhiên chỉ để che tập rỗng.")
```

### 13.1. Tạo pipeline biến đổi — fit chỉ trên train

Cell này xây dựng pipeline bằng scikit-learn. `preprocessor.fit(...)` chỉ nhận X của tập train; validation/test không được dùng để tính median, trung bình/độ lệch chuẩn hoặc danh sách giá trị phân loại. Để tránh ngốn RAM khi dữ liệu lên vài triệu dòng, mặc định chỉ biến đổi thử 10.000 mẫu train để kiểm tra pipeline. Có thể bật `TRANSFORM_ALL_SPLITS = True` nếu cần vật hóa toàn bộ ma trận đặc trưng và máy đủ RAM.

```python
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder, StandardScaler

numeric_pipeline = Pipeline(steps=[
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler()),
])
categorical_pipeline = Pipeline(steps=[
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("onehot", OneHotEncoder(handle_unknown="ignore", min_frequency=20, sparse_output=True)),
])
preprocessor = ColumnTransformer(
    transformers=[
        ("num", numeric_pipeline, MODEL_NUMERIC_FINAL),
        ("cat", categorical_pipeline, MODEL_CATEGORICAL_FINAL),
    ],
    remainder="drop",
    sparse_threshold=1.0,
)

X_train = train_df[MODEL_FEATURES_FINAL]
y_train = train_df[TARGET].astype("float32")
X_valid = valid_df[MODEL_FEATURES_FINAL]
y_valid = valid_df[TARGET].astype("float32")
X_test = test_df[MODEL_FEATURES_FINAL]
y_test = test_df[TARGET].astype("float32")

if len(X_train):
    preprocessor.fit(X_train)  # Fit duy nhất trên train.
    print("Đã fit phép biến đổi trên tập train.")
    print("Train X/y:", X_train.shape, y_train.shape)
    print("Validation X/y:", X_valid.shape, y_valid.shape)
    print("Test X/y:", X_test.shape, y_test.shape)

    TRANSFORM_ALL_SPLITS = False
    PREVIEW_ROWS = min(10_000, len(X_train))
    X_train_preview = preprocessor.transform(X_train.iloc[:PREVIEW_ROWS])
    print(f"Kích thước ma trận train sau biến đổi thử ({PREVIEW_ROWS:,} dòng): {X_train_preview.shape}")
    print("Dạng ma trận:", type(X_train_preview).__name__, "| số đặc trưng sau One-Hot:", X_train_preview.shape[1])

    if TRANSFORM_ALL_SPLITS:
        X_train_transformed = preprocessor.transform(X_train)
        X_valid_transformed = preprocessor.transform(X_valid)
        X_test_transformed = preprocessor.transform(X_test)
        print("Đã biến đổi toàn bộ train/validation/test. Lưu ý kiểm soát RAM khi chạy dữ liệu nhiều triệu dòng.")
else:
    print("Không thể fit pipeline vì tập train rỗng. Hãy kiểm tra lại mốc thời gian chia tập.")
```

## 14. Chiến lược đánh giá và chỉ số phù hợp

Vì nhãn `ARR_DELAY` là số phút nên đây là **bài toán hồi quy**:

- **MAE (Mean Absolute Error):** sai số tuyệt đối trung bình, dễ diễn giải thành số phút dự đoán lệch trung bình bao nhiêu.
- **RMSE (Root Mean Squared Error):** phạt sai số lớn mạnh hơn MAE; hữu ích vì các chuyến trễ rất lâu có thể gây ảnh hưởng lớn.
- **R²:** đo mức độ biến thiên của nhãn được giải thích so với dự đoán bằng trung bình; có thể âm nếu mô hình dự đoán tệ hơn baseline.

Dùng các tháng cuối làm test mô phỏng tình huống huấn luyện bằng dữ liệu đã có rồi dự đoán tháng tương lai. Validation dùng để chọn tham số/mô hình; test chỉ dùng đánh giá cuối cùng. Khi có đủ dữ liệu và cần đánh giá ổn định hơn, có thể bổ sung backtesting theo nhiều cửa sổ thời gian (walk-forward/TimeSeriesSplit). Không dùng Accuracy/F1 làm chỉ số chính vì đó là chỉ số phân loại.

Notebook này chưa huấn luyện mô hình cuối; bước tiếp theo có thể so sánh baseline dự đoán median với các mô hình hồi quy như SGDRegressor, Ridge hoặc LightGBM, nhưng phải dùng cùng cách chia theo thời gian và cùng bộ chỉ số.

```python
# Baseline tham khảo không dùng X: luôn dự đoán median ARR_DELAY của train.
# Đây là đường chuẩn đơn giản để sau này so sánh mô hình, không phải mô hình cuối cùng.
if len(y_train) and len(y_valid) and len(y_test):
    from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
    baseline_value = float(y_train.median())
    baseline_rows = []
    for split_name, y_true in [("Validation", y_valid), ("Test", y_test)]:
        pred = np.full(len(y_true), baseline_value, dtype="float32")
        baseline_rows.append({
            "Tập": split_name,
            "Giá trị dự đoán baseline (median train, phút)": baseline_value,
            "MAE (phút)": mean_absolute_error(y_true, pred),
            "RMSE (phút)": np.sqrt(mean_squared_error(y_true, pred)),
            "R2": r2_score(y_true, pred) if len(y_true) > 1 else np.nan,
        })
    baseline_df = pd.DataFrame(baseline_rows)
    display(baseline_df.round(4))
    baseline_df.to_csv(TABLES_DIR / "18_baseline_median_train.csv", index=False, encoding="utf-8-sig")
    print("Khi huấn luyện mô hình, cần cải thiện baseline trên validation; chỉ đánh giá test sau khi đã chốt mô hình/tham số.")
else:
    print("Không tính baseline vì một trong các tập train/validation/test đang rỗng.")
```
