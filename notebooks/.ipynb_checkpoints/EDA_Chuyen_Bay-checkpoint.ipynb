Notebook được trình bày theo quy trình khảo sát dữ liệu, có bảng, biểu đồ và nhận xét được sinh tự động dựa trên dữ liệu thực tế. Notebook không tự ý xóa ngoại lai hay bản ghi; các điểm bất thường được **định lượng và đánh dấu** để người làm mô hình quyết định cách xử lý.

## Cách sử dụng và đóng gói để nộp

1. Đặt notebook ở **thư mục gốc của dự án**, cùng cấp với thư mục `data/`.
2. Giữ cấu trúc `data/raw/` cho dữ liệu chưa xử lý và `data/ouput/` cho dữ liệu đã xử lý. Notebook cũng tự tìm `data/output/` hoặc `data/processed/` nếu tên thư mục của bạn khác.
3. Trong Jupyter, chọn **Run All** một lần; kiểm tra kết quả, sau đó **Save** notebook (`Ctrl+S`).
4. Nén thư mục dự án, gồm notebook và dữ liệu cần thiết, thành `.zip` để gửi thầy. Khi notebook đã được chạy và lưu, biểu đồ/bảng/nhận xét sẽ được lưu trong file `.ipynb`; thầy có thể mở để xem mà không cần chạy lại. Nếu thầy muốn tự kiểm chứng hoặc tái tạo kết quả thì vẫn có thể chạy lại.

> **Lưu ý về nhãn:** ví dụ dữ liệu bạn cung cấp có `ARR_DELAY` là số phút trễ đến (nhãn liên tục, phù hợp bài toán hồi quy). Để đáp ứng mục phân bố nhãn/mất cân bằng lớp, nếu dữ liệu không có nhãn phân loại sẵn, notebook tạo thêm nhãn phân tích `DELAYED_15M` = 1 khi `ARR_DELAY > 15` phút và = 0 trong trường hợp còn lại. Đây là quy ước phân tích có thể chỉnh ở phần cấu hình, **không ghi đè dữ liệu gốc**.

**Giới hạn hiện tại:** notebook này được tạo từ cấu trúc cột mẫu bạn gửi, nhưng trong cuộc trò chuyện chưa có tệp dữ liệu thực tế để chạy và lưu sẵn kết quả. Sau khi đặt dữ liệu vào đúng thư mục và chạy một lần, cần lưu notebook để các kết quả thực tế được nhúng trong file nộp.

```python
# 1. CẤU HÌNH VÀ THƯ VIỆN
from pathlib import Path
import re
import warnings
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from IPython.display import display, Markdown

warnings.filterwarnings("ignore", category=FutureWarning)
pd.set_option("display.max_columns", 100)
pd.set_option("display.max_rows", 100)
pd.set_option("display.width", 160)

# Notebook nên được mở/chạy từ thư mục gốc của dự án.
PROJECT_DIR = Path.cwd()
RAW_DIR = PROJECT_DIR / "data" / "raw"
PROCESSED_CANDIDATES = [
    PROJECT_DIR / "data" / "ouput",      # tên thư mục theo dự án hiện tại
    PROJECT_DIR / "data" / "output",     # tên viết đúng phổ biến
    PROJECT_DIR / "data" / "processed",
]
PROCESSED_DIR = next((p for p in PROCESSED_CANDIDATES if p.exists()), PROCESSED_CANDIDATES[0])
REPORT_DIR = PROJECT_DIR / "reports"
FIGURE_DIR = REPORT_DIR / "figures"
FIGURE_DIR.mkdir(parents=True, exist_ok=True)

# Nếu dữ liệu không nằm ở data/raw hoặc data/ouput, hãy sửa 2 biến RAW_DIR / PROCESSED_DIR phía trên.
# Ngưỡng phân loại trễ (phút), chỉ dùng khi không có nhãn phân loại sẵn.
DELAY_THRESHOLD_MINUTES = 15
MAX_PLOT_ROWS = 100_000       # chỉ lấy mẫu khi vẽ biểu đồ để tránh biểu đồ quá nặng
RANDOM_STATE = 42

print("Thư mục dự án:", PROJECT_DIR)
print("Dữ liệu thô:", RAW_DIR, "| tồn tại:", RAW_DIR.exists())
print("Dữ liệu đã xử lý:", PROCESSED_DIR, "| tồn tại:", PROCESSED_DIR.exists())
print("Thư mục hình xuất ra:", FIGURE_DIR)
```

## 2. Nạp dữ liệu thô và dữ liệu đã xử lý

Phần này đọc các tệp CSV/CSV nén, Parquet hoặc Excel trong hai thư mục; đồng thời lưu tên tệp nguồn để hỗ trợ truy vết. Không thay đổi hay ghi đè các tệp đầu vào. Dữ liệu đã xử lý được ưu tiên cho EDA; dữ liệu thô được dùng để đối chiếu chất lượng trước và sau xử lý.

```python
SUPPORTED_SUFFIXES = (".csv", ".csv.gz", ".parquet", ".xlsx", ".xls")

def list_data_files(folder: Path):
    if not folder.exists():
        return []
    files = []
    for p in folder.rglob("*"):
        if not p.is_file() or p.name.startswith("~$") or p.name.startswith("."):
            continue
        name = p.name.lower()
        if name.endswith(SUPPORTED_SUFFIXES):
            files.append(p)
    return sorted(files)

def read_one_file(path: Path) -> pd.DataFrame:
    suffix = path.name.lower()
    if suffix.endswith(".parquet"):
        return pd.read_parquet(path)
    if suffix.endswith((".xlsx", ".xls")):
        return pd.read_excel(path)
    # CSV thường dùng dấu phẩy; nhận diện thêm trường hợp dùng dấu chấm phẩy.
    try:
        frame = pd.read_csv(path, low_memory=False, encoding="utf-8")
    except UnicodeDecodeError:
        frame = pd.read_csv(path, low_memory=False, encoding="latin-1")
    if frame.shape[1] == 1:
        try:
            first_line = path.open("r", encoding="utf-8", errors="ignore").readline()
            if ";" in first_line and "," not in first_line:
                frame = pd.read_csv(path, sep=";", low_memory=False, encoding="utf-8")
        except Exception:
            pass
    return frame

def load_folder(folder: Path, label: str):
    files = list_data_files(folder)
    frames, log = [], []
    for path in files:
        try:
            part = read_one_file(path)
            part.columns = [str(c).strip() for c in part.columns]
            part["_SOURCE_FILE"] = str(path.relative_to(PROJECT_DIR)) if path.is_relative_to(PROJECT_DIR) else str(path)
            frames.append(part)
            log.append({"Loại dữ liệu": label, "Tệp": path.name, "Số dòng": len(part), "Số cột (kể cả nguồn)": part.shape[1]})
        except Exception as exc:
            log.append({"Loại dữ liệu": label, "Tệp": path.name, "Số dòng": np.nan, "Số cột (kể cả nguồn)": np.nan, "Lỗi đọc": str(exc)[:250]})
            print(f"CẢNH BÁO: không đọc được {path}: {exc}")
    combined = pd.concat(frames, ignore_index=True, sort=False) if frames else pd.DataFrame()
    return combined, log, files

raw_df, raw_log, raw_files = load_folder(RAW_DIR, "Raw")
processed_df, processed_log, processed_files = load_folder(PROCESSED_DIR, "Đã xử lý")
file_log = pd.DataFrame(raw_log + processed_log)

if not file_log.empty:
    display(file_log)
else:
    display(Markdown("**Chưa tìm thấy tệp dữ liệu hỗ trợ.** Hãy đặt các tệp CSV/Parquet/Excel vào `data/raw/` và `data/ouput/`, rồi chạy lại notebook."))

if processed_df.empty and raw_df.empty:
    raise FileNotFoundError(
        "Không tìm thấy dữ liệu. Hãy đặt dữ liệu thô ở data/raw/ và dữ liệu đã xử lý ở data/ouput/ "
        "(hoặc sửa RAW_DIR/PROCESSED_DIR trong phần cấu hình)."
    )

# Ưu tiên dữ liệu đã xử lý cho EDA; nếu thư mục output trống thì tạm khảo sát dữ liệu raw.
using_processed_data = not processed_df.empty
df_source = processed_df if using_processed_data else raw_df
df = df_source.copy()
df.columns = [str(c).strip() for c in df.columns]

# Một vài cột có thể được đọc thành chuỗi do kiểu dữ liệu không đồng nhất.
# Không ép kiểu toàn bộ ở đây để bảng chất lượng có thể phát hiện lỗi định dạng.
print(f"Nguồn dùng cho EDA: {'đã xử lý' if using_processed_data else 'raw (fallback vì chưa có dữ liệu đã xử lý)'}")
print(f"Kích thước dữ liệu EDA: {df.shape[0]:,} dòng × {df.shape[1]:,} cột")
print("Tên tệp:", len(processed_files if using_processed_data else raw_files))
```

## 3. Khám phá tổng quan và thống kê mô tả

Xem kích thước, kiểu dữ liệu, ví dụ bản ghi và thống kê mô tả. Với dữ liệu đã xử lý, các cột dự kiến gồm `MONTH`, `DAY_OF_MONTH`, `DAY_OF_WEEK`, `OP_UNIQUE_CARRIER`, `ORIGIN`, `DEST`, các giờ bay theo lịch, `ARR_DELAY`, thời gian bay dự kiến, khoảng cách và thời tiết tại sân bay đi/đến.

```python
display(Markdown("### 3.1. Kích thước và mẫu bản ghi"))
display(pd.DataFrame({
    "Chỉ tiêu": ["Số dòng", "Số cột", "Số tệp dữ liệu", "Số dòng trùng toàn bộ (ước tính sơ bộ)"],
    "Giá trị": [len(df), len(df.columns), len(processed_files if using_processed_data else raw_files), int(df.drop(columns=["_SOURCE_FILE"], errors="ignore").duplicated().sum())]
}))
display(df.head(10))

display(Markdown("### 3.2. Kiểu dữ liệu và số lượng giá trị khác nhau"))
schema = pd.DataFrame({
    "Cột": df.columns,
    "Kiểu dữ liệu": [str(df[c].dtype) for c in df.columns],
    "Số giá trị khác nhau": [df[c].nunique(dropna=True) for c in df.columns],
    "Số giá trị thiếu": [int(df[c].isna().sum()) for c in df.columns],
    "Tỷ lệ thiếu (%)": [round(df[c].isna().mean()*100, 3) for c in df.columns],
})
display(schema.sort_values(["Tỷ lệ thiếu (%)", "Cột"], ascending=[False, True]))

display(Markdown("### 3.3. Thống kê mô tả các biến số"))
numeric_cols_all = df.select_dtypes(include=np.number).columns.tolist()
if numeric_cols_all:
    display(df[numeric_cols_all].describe(percentiles=[.01, .05, .25, .5, .75, .95, .99]).T)
else:
    display(Markdown("Không có cột nào đang được nhận diện là kiểu số. Hãy kiểm tra phần định dạng dữ liệu."))

display(Markdown("### Nhận xét tổng quan"))
non_meta_cols = [c for c in df.columns if c != "_SOURCE_FILE"]
missing_cells = int(df[non_meta_cols].isna().sum().sum()) if non_meta_cols else 0
all_cells = int(len(df) * len(non_meta_cols))
missing_pct = 100 * missing_cells / all_cells if all_cells else 0
summary_notes = [
    f"- Tập EDA có **{len(df):,} dòng** và **{len(non_meta_cols)} cột dữ liệu** (không tính cột truy vết tệp).",
    f"- Tổng tỷ lệ ô thiếu trên các cột dữ liệu là **{missing_pct:.2f}%** ({missing_cells:,}/{all_cells:,} ô).",
    f"- Có **{len(numeric_cols_all)} cột kiểu số** và **{len(df.columns)-len(numeric_cols_all)} cột không phải kiểu số** theo pandas; kiểu dữ liệu cần được đối chiếu với ý nghĩa từng trường."
]
display(Markdown("\n".join(summary_notes)))
```

## 4. Khảo sát thống kê và phân bố dữ liệu (tiêu chí 4 – 1,0 điểm)

Các biểu đồ dưới đây sử dụng dữ liệu đã xử lý (nếu có). Mỗi biểu đồ có nhận xét tự động ngay bên dưới. Nếu dữ liệu rất lớn, notebook lấy mẫu tối đa `MAX_PLOT_ROWS` dòng để vẽ; bảng thống kê chất lượng vẫn tính trên toàn bộ dữ liệu đã nạp.

```python
# Chuẩn hóa tên cột để tìm trường không phụ thuộc chữ hoa/thường.
col_lookup = {str(c).strip().upper(): c for c in df.columns}
def get_col(*names):
    for name in names:
        if name.upper() in col_lookup:
            return col_lookup[name.upper()]
    return None

def sample_for_plot(frame, max_rows=MAX_PLOT_ROWS):
    if len(frame) > max_rows:
        return frame.sample(max_rows, random_state=RANDOM_STATE)
    return frame

plot_df = sample_for_plot(df)

def save_and_show(fig, filename):
    out = FIGURE_DIR / filename
    fig.savefig(out, dpi=150, bbox_inches="tight")
    plt.show()
    plt.close(fig)
    print(f"Đã lưu hình: {out.relative_to(PROJECT_DIR) if out.is_relative_to(PROJECT_DIR) else out}")

def comment(text):
    display(Markdown(f"**Nhận xét:** {text}"))

# 4.1 Nhãn đầu ra dạng liên tục và nhãn phân loại (nếu có thể xác định).
arr_delay_col = get_col("ARR_DELAY")
existing_label_col = get_col("ARR_DEL15", "DELAYED_15M", "IS_DELAYED", "DELAY_LABEL", "ARR_DELAY_CLASS", "LABEL", "TARGET")
analysis_df = df.copy()
label_col = None
label_created = False

if existing_label_col is not None:
    label_col = existing_label_col
    print(f"Dùng nhãn phân loại có sẵn: {label_col}")
elif arr_delay_col is not None:
    analysis_df["DELAYED_15M_ANALYSIS"] = pd.to_numeric(analysis_df[arr_delay_col], errors="coerce").gt(DELAY_THRESHOLD_MINUTES).map({True: "Trễ trên 15 phút", False: "Không trễ trên 15 phút"})
    label_col = "DELAYED_15M_ANALYSIS"
    label_created = True
    print(f"Không tìm thấy nhãn phân loại sẵn; tạo nhãn phân tích ARR_DELAY > {DELAY_THRESHOLD_MINUTES} phút. Đây không phải cột gốc.")
else:
    print("Không tìm thấy ARR_DELAY hoặc nhãn phân loại quen thuộc; sẽ bỏ qua biểu đồ nhãn.")

# Biểu đồ phân bố của ARR_DELAY, là đầu ra liên tục trong ví dụ dữ liệu người dùng cung cấp.
if arr_delay_col is not None:
    x = pd.to_numeric(plot_df[arr_delay_col], errors="coerce").dropna()
    if not x.empty:
        fig, ax = plt.subplots(figsize=(10, 5))
        ax.hist(x.clip(lower=x.quantile(.01), upper=x.quantile(.99)), bins=50, edgecolor="white")
        ax.set_title("Phân bố ARR_DELAY (cắt hiển thị ở phân vị 1%–99%)")
        ax.set_xlabel("ARR_DELAY (phút; âm = đến sớm, dương = đến trễ)")
        ax.set_ylabel("Số chuyến bay")
        ax.grid(axis="y", alpha=.25)
        save_and_show(fig, "01_phan_bo_arr_delay.png")
        med, mean, q90 = x.median(), x.mean(), x.quantile(.90)
        skew_desc = "lệch phải" if x.skew() > .3 else ("lệch trái" if x.skew() < -.3 else "khá cân đối")
        comment(f"ARR_DELAY có trung vị **{med:.2f} phút**, trung bình **{mean:.2f} phút** và phân vị 90% là **{q90:.2f} phút**; phân bố **{skew_desc}** (skewness = {x.skew():.2f}). Các giá trị âm biểu thị chuyến đến sớm; phần đuôi xa cần được kiểm tra như ngoại lai tiềm năng, không tự động xóa.")

# Biểu đồ phân bố nhãn phân loại.
if label_col is not None:
    lbl = analysis_df[label_col].dropna().astype(str)
    counts = lbl.value_counts()
    if not counts.empty:
        fig, ax = plt.subplots(figsize=(8, 4.5))
        counts.plot(kind="bar", ax=ax, edgecolor="white")
        ax.set_title(f"Phân bố nhãn: {label_col}")
        ax.set_xlabel("Nhãn")
        ax.set_ylabel("Số chuyến bay")
        ax.tick_params(axis="x", rotation=20)
        ax.grid(axis="y", alpha=.25)
        save_and_show(fig, "02_phan_bo_nhan.png")
        pcts = (counts / counts.sum() * 100).round(2)
        comment("Phân bố nhãn: " + "; ".join([f"**{k}** = {v:,} chuyến ({pcts[k]:.2f}%)" for k, v in counts.items()]) + ". " + ("Tỷ lệ lớp lớn nhất/lớp nhỏ nhất cao hơn 3 lần, cần cân nhắc stratified split, class weights hoặc metric phù hợp nếu đây là bài toán phân loại." if len(counts) >= 2 and counts.max()/max(counts.min(), 1) > 3 else "Chưa thấy chênh lệch trên 3 lần giữa lớp lớn nhất và nhỏ nhất theo thống kê đếm này; vẫn nên kiểm tra phân bố trên tập train/test."))

# Chọn các biến số có ý nghĩa để vẽ phân bố.
preferred_numeric = [
    "DISTANCE", "CRS_ELAPSED_TIME", "ORIGIN_TEMP_C", "DEST_TEMP_C",
    "ORIGIN_PRECIPITATION_MM", "DEST_PRECIPITATION_MM",
    "ORIGIN_WIND_SPEED_10M_KMH", "DEST_WIND_SPEED_10M_KMH"
]
selected_num = []
for nm in preferred_numeric:
    c = get_col(nm)
    if c and c not in selected_num:
        selected_num.append(c)
# Bổ sung các biến số khác nếu danh sách trên không đủ.
for c in plot_df.select_dtypes(include=np.number).columns:
    if c not in selected_num and c not in ["_SOURCE_FILE"] and len(selected_num) < 8:
        if c not in [get_col("MONTH"), get_col("DAY_OF_MONTH"), get_col("DAY_OF_WEEK"), get_col("CRS_DEP_TIME"), get_col("CRS_ARR_TIME")]:
            selected_num.append(c)

for idx, c in enumerate(selected_num, start=1):
    x = pd.to_numeric(plot_df[c], errors="coerce").dropna()
    if x.empty or x.nunique() <= 1:
        continue
    low, high = x.quantile(.01), x.quantile(.99)
    shown = x.clip(lower=low, upper=high)
    fig, ax = plt.subplots(figsize=(8, 4))
    ax.hist(shown, bins=35, edgecolor="white")
    ax.set_title(f"Phân bố đặc trưng: {c}")
    ax.set_xlabel(c)
    ax.set_ylabel("Số bản ghi")
    ax.grid(axis="y", alpha=.25)
    fname = f"03_phan_bo_{idx:02d}_{re.sub(r'[^A-Za-z0-9]+', '_', str(c))[:30]}.png"
    save_and_show(fig, fname)
    comment(f"{c}: trung vị = **{x.median():.3f}**, trung bình = **{x.mean():.3f}**, khoảng phân vị 5%–95% = **[{x.quantile(.05):.3f}, {x.quantile(.95):.3f}]**. Hình hiển thị được giới hạn ở phân vị 1%–99% để thấy rõ phần phân bố chính; các giá trị cực trị vẫn được giữ trong dữ liệu và được định lượng ở mục chất lượng dữ liệu.")

# Các biến phân loại/biến thời gian.
preferred_cats = ["OP_UNIQUE_CARRIER", "MONTH", "DAY_OF_WEEK", "ORIGIN", "DEST"]
selected_cat = []
for nm in preferred_cats:
    c = get_col(nm)
    if c and c not in selected_cat:
        selected_cat.append(c)
for c in df.columns:
    if c in selected_cat or c == "_SOURCE_FILE":
        continue
    if (df[c].dtype == "object" or str(df[c].dtype) == "category") and df[c].nunique(dropna=True) <= 50 and len(selected_cat) < 7:
        selected_cat.append(c)

for idx, c in enumerate(selected_cat, start=1):
    counts = df[c].astype("string").fillna("(Thiếu)").value_counts().head(15)
    if counts.empty:
        continue
    fig, ax = plt.subplots(figsize=(9, 4.5))
    counts.sort_values().plot(kind="barh", ax=ax, edgecolor="white")
    ax.set_title(f"Top nhóm/tần suất: {c}" + (" (15 nhóm phổ biến nhất)" if df[c].nunique(dropna=True) > 15 else ""))
    ax.set_xlabel("Số bản ghi")
    ax.set_ylabel(c)
    ax.grid(axis="x", alpha=.25)
    save_and_show(fig, f"04_phan_bo_nhom_{idx:02d}_{re.sub(r'[^A-Za-z0-9]+', '_', str(c))[:25]}.png")
    most = counts.idxmax()
    share = counts.max() / max(counts.sum(), 1) * 100
    comment(f"Ở biến **{c}**, nhóm xuất hiện nhiều nhất trong các nhóm đang hiển thị là **{most}** ({counts.max():,} bản ghi; khoảng {share:.2f}% tổng tần suất trong hình). Các nhóm có tần suất thấp có thể tạo ước lượng kém ổn định; với `ORIGIN`/`DEST` chỉ hiển thị tối đa 15 nhóm nhiều nhất để biểu đồ dễ đọc.")
```

## 5. Phân tích quan hệ giữa các đặc trưng và đầu ra (tiêu chí 5 – 1,0 điểm)

Phân tích tương quan chỉ cho biết mức độ liên hệ thống kê, **không chứng minh quan hệ nhân quả**. Với dữ liệu chuyến bay, giờ theo lịch và ngày trong tuần có tính chu kỳ; vì vậy hệ số Pearson của các mã giờ dạng `HHMM` chỉ mang tính tham khảo và không nên diễn giải như thời gian liên tục thông thường.

```python
# 5.1 Ma trận tương quan giữa các biến số hợp lý.
exclude_from_corr = {get_col("MONTH"), get_col("DAY_OF_MONTH"), get_col("DAY_OF_WEEK"), get_col("CRS_DEP_TIME"), get_col("CRS_ARR_TIME"), "_SOURCE_FILE"}
valid_numeric = []
for c in df.select_dtypes(include=np.number).columns:
    if c not in exclude_from_corr:
        valid_numeric.append(c)
# Thêm các trường thường được đọc dạng chuỗi nhưng thực chất là số, khi chuyển đổi được.
for nm in ["ARR_DELAY", "DISTANCE", "CRS_ELAPSED_TIME", "ORIGIN_TEMP_C", "DEST_TEMP_C", "ORIGIN_PRECIPITATION_MM", "DEST_PRECIPITATION_MM", "ORIGIN_WIND_SPEED_10M_KMH", "DEST_WIND_SPEED_10M_KMH"]:
    c = get_col(nm)
    if c and c not in valid_numeric:
        converted = pd.to_numeric(df[c], errors="coerce")
        if converted.notna().sum() > 0:
            df_for_corr = df.copy()
            df_for_corr[c] = converted
            valid_numeric.append(c)
        else:
            df_for_corr = df.copy()
else:
    df_for_corr = df.copy()
# Chuyển các biến đã chọn sang numeric; các giá trị lỗi định dạng thành NaN cho riêng phép tính tương quan.
for c in valid_numeric:
    df_for_corr[c] = pd.to_numeric(df_for_corr[c], errors="coerce")
valid_numeric = [c for c in valid_numeric if df_for_corr[c].notna().sum() > 2 and df_for_corr[c].nunique(dropna=True) > 1]
if len(valid_numeric) >= 2:
    corr = df_for_corr[valid_numeric].corr(numeric_only=True)
    fig, ax = plt.subplots(figsize=(max(8, .85*len(corr.columns)), max(6, .7*len(corr.columns))))
    im = ax.imshow(corr.values, vmin=-1, vmax=1, cmap="coolwarm", aspect="auto")
    ax.set_xticks(range(len(corr.columns)))
    ax.set_xticklabels(corr.columns, rotation=70, ha="right", fontsize=8)
    ax.set_yticks(range(len(corr.index)))
    ax.set_yticklabels(corr.index, fontsize=8)
    ax.set_title("Ma trận tương quan Pearson giữa các biến số")
    fig.colorbar(im, ax=ax, fraction=.025, pad=.03, label="Hệ số tương quan")
    save_and_show(fig, "05_ma_tran_tuong_quan.png")
    comment("Các ô gần **+1** thể hiện quan hệ tuyến tính cùng chiều, gần **−1** thể hiện ngược chiều, gần **0** thể hiện ít quan hệ tuyến tính. Tương quan thấp không có nghĩa là không có quan hệ phi tuyến; tương quan cao giữa hai biến đầu vào cũng có thể báo hiệu đa cộng tuyến.")
else:
    corr = pd.DataFrame()
    display(Markdown("Không đủ biến số có phương sai để lập ma trận tương quan."))

# 5.2 Xếp hạng tương quan với ARR_DELAY (nếu có).
if arr_delay_col is not None and not corr.empty and arr_delay_col in corr.columns:
    target_corr = corr[arr_delay_col].drop(labels=[arr_delay_col], errors="ignore").dropna()
    target_corr = target_corr.reindex(target_corr.abs().sort_values(ascending=False).index)
    display(Markdown("### Tương quan của các đặc trưng số với ARR_DELAY"))
    display(target_corr.rename("Pearson r với ARR_DELAY").to_frame().round(3))
    if not target_corr.empty:
        top = target_corr.head(1)
        first_col, first_val = top.index[0], float(top.iloc[0])
        comment(f"Trong các biến số được đưa vào phép tính, **{first_col}** có tương quan tuyến tính mạnh nhất về trị tuyệt đối với `ARR_DELAY` (r = **{first_val:.3f}**). Đây là ứng viên cần khảo sát thêm, không đồng nghĩa biến đó gây ra chậm chuyến.")

# 5.3 Quan hệ nhóm với ARR_DELAY: hãng bay, khoảng cách, thời tiết.
if arr_delay_col is not None:
    rel_df = df.copy()
    rel_df["_ARR_DELAY_NUM"] = pd.to_numeric(rel_df[arr_delay_col], errors="coerce")
    carrier_col = get_col("OP_UNIQUE_CARRIER")
    if carrier_col:
        carrier_stats = rel_df.groupby(carrier_col, dropna=False)["_ARR_DELAY_NUM"].agg(["count", "mean", "median"]).query("count >= 20").sort_values("mean", ascending=False)
        if not carrier_stats.empty:
            display(Markdown("### ARR_DELAY theo hãng bay (chỉ nhóm có ít nhất 20 bản ghi)"))
            display(carrier_stats.round(2))
            shown = pd.concat([carrier_stats.head(8), carrier_stats.tail(8)]).drop_duplicates().sort_values("mean")
            fig, ax = plt.subplots(figsize=(9, max(4, .35*len(shown))))
            shown["mean"].plot(kind="barh", ax=ax, edgecolor="white")
            ax.set_title("Trung bình ARR_DELAY theo hãng bay (các nhóm cao/thấp)")
            ax.set_xlabel("ARR_DELAY trung bình (phút)")
            ax.set_ylabel("Hãng bay")
            ax.grid(axis="x", alpha=.25)
            save_and_show(fig, "06_arr_delay_theo_hang_bay.png")
            high = carrier_stats.iloc[0]
            low = carrier_stats.iloc[-1]
            comment(f"Trong nhóm có ít nhất 20 chuyến, hãng **{carrier_stats.index[0]}** có trung bình ARR_DELAY lớn nhất ({high['mean']:.2f} phút), còn **{carrier_stats.index[-1]}** thấp nhất ({low['mean']:.2f} phút). Cần xét cỡ mẫu, tuyến bay và mùa vụ trước khi kết luận; đây là quan sát mô tả.")

    distance_col = get_col("DISTANCE")
    if distance_col:
        rel_df["_DISTANCE_NUM"] = pd.to_numeric(rel_df[distance_col], errors="coerce")
        usable = rel_df[["_ARR_DELAY_NUM", "_DISTANCE_NUM"]].dropna()
        if len(usable) >= 20 and usable["_DISTANCE_NUM"].nunique() >= 4:
            try:
                usable["Nhóm khoảng cách"] = pd.qcut(usable["_DISTANCE_NUM"], q=min(5, usable["_DISTANCE_NUM"].nunique()), duplicates="drop")
                dist_stats = usable.groupby("Nhóm khoảng cách", observed=False)["_ARR_DELAY_NUM"].agg(["count", "mean", "median"]).dropna()
                display(Markdown("### ARR_DELAY theo nhóm khoảng cách (phân vị)"))
                display(dist_stats.round(2))
                fig, ax = plt.subplots(figsize=(9, 4.5))
                dist_stats["mean"].plot(kind="bar", ax=ax, edgecolor="white")
                ax.set_title("ARR_DELAY trung bình theo nhóm khoảng cách")
                ax.set_ylabel("ARR_DELAY trung bình (phút)")
                ax.set_xlabel("Nhóm khoảng cách (phân vị)")
                ax.tick_params(axis="x", rotation=25)
                ax.grid(axis="y", alpha=.25)
                save_and_show(fig, "07_arr_delay_theo_khoang_cach.png")
                comment("So sánh trung bình/ trung vị giữa các nhóm khoảng cách để kiểm tra liệu hành vi chậm chuyến có khác nhau theo quãng đường hay không. Nếu các nhóm không tăng/giảm đều, quan hệ có thể phi tuyến hoặc bị ảnh hưởng bởi tuyến bay, sân bay và thời điểm.")
            except Exception as exc:
                print("Không lập được nhóm khoảng cách:", exc)

    # Thời tiết: tương quan Pearson và nhóm mưa nếu biến có đủ độ biến thiên.
    weather_cols = [get_col("ORIGIN_PRECIPITATION_MM"), get_col("DEST_PRECIPITATION_MM"), get_col("ORIGIN_WIND_SPEED_10M_KMH"), get_col("DEST_WIND_SPEED_10M_KMH"), get_col("ORIGIN_TEMP_C"), get_col("DEST_TEMP_C")]
    weather_cols = [c for c in weather_cols if c]
    weather_rows = []
    for c in weather_cols:
        vals = pd.to_numeric(rel_df[c], errors="coerce")
        valid = pd.concat([rel_df["_ARR_DELAY_NUM"], vals.rename("_weather")], axis=1).dropna()
        if len(valid) >= 10 and valid["_weather"].nunique() > 1:
            weather_rows.append({"Đặc trưng thời tiết": c, "Pearson r với ARR_DELAY": valid["_weather"].corr(valid["_ARR_DELAY_NUM"]), "Số dòng hợp lệ": len(valid)})
    if weather_rows:
        weather_corr = pd.DataFrame(weather_rows).sort_values("Pearson r với ARR_DELAY", key=lambda s: s.abs(), ascending=False)
        display(Markdown("### Quan hệ tuyến tính sơ bộ giữa thời tiết và ARR_DELAY"))
        display(weather_corr.round(3))
        best = weather_corr.iloc[0]
        comment(f"Trong các biến thời tiết có sẵn, **{best['Đặc trưng thời tiết']}** có tương quan tuyến tính mạnh nhất về trị tuyệt đối với ARR_DELAY (r = {best['Pearson r với ARR_DELAY']:.3f}, n = {int(best['Số dòng hợp lệ']):,}). Hệ số nhỏ cho thấy thời tiết đơn lẻ có thể chưa giải thích nhiều biến động; nên đánh giá cùng tuyến bay, thời gian và các yếu tố vận hành.")
else:
    display(Markdown("Không có `ARR_DELAY`, nên bỏ qua phần quan hệ với đầu ra liên tục. Hãy xác nhận tên cột mục tiêu nếu nó có tên khác."))
```

### 5.4. Tóm tắt ít nhất 3 nhận xét có ý nghĩa cho xây dựng mô hình

Các nhận xét dưới đây được sinh từ số liệu thực tế đang nạp vào notebook. Khi dùng trong báo cáo, cần kiểm tra chúng cùng với ngữ cảnh nghiệp vụ và cách chia train/validation/test.

```python
findings = []
# Nhận xét 1: tương quan mục tiêu.
if arr_delay_col is not None and not corr.empty and arr_delay_col in corr.columns:
    tc = corr[arr_delay_col].drop(labels=[arr_delay_col], errors="ignore").dropna()
    if not tc.empty:
        tc = tc.reindex(tc.abs().sort_values(ascending=False).index)
        c, r = tc.index[0], float(tc.iloc[0])
        findings.append(f"**Quan hệ đặc trưng–nhãn:** `{c}` có tương quan Pearson lớn nhất về trị tuyệt đối với `ARR_DELAY` trong các biến số được xét (r = {r:.3f}). Nên so sánh mô hình có/không có biến này và kiểm tra tính sẵn có của biến tại thời điểm dự báo.")
# Nhận xét 2: khác biệt theo hãng.
if arr_delay_col is not None and carrier_col if 'carrier_col' in globals() else False:
    if 'carrier_stats' in globals() and not carrier_stats.empty:
        findings.append(f"**Khác biệt nhóm:** trung bình ARR_DELAY khác nhau giữa các hãng trong dữ liệu (cao nhất {carrier_stats.iloc[0]['mean']:.2f} phút ở {carrier_stats.index[0]}, thấp nhất {carrier_stats.iloc[-1]['mean']:.2f} phút ở {carrier_stats.index[-1]}, chỉ xét nhóm có ≥20 chuyến). Có thể mã hóa hãng/tuyến theo cách phù hợp, đồng thời tránh để nhóm hiếm tạo ước lượng quá nhiễu.")
# Nhận xét 3: khoảng cách.
if arr_delay_col is not None and 'dist_stats' in globals() and not dist_stats.empty:
    means = dist_stats['mean'].dropna()
    if len(means) >= 2:
        findings.append(f"**Quan hệ với khoảng cách:** ARR_DELAY trung bình theo nhóm khoảng cách dao động từ {means.min():.2f} đến {means.max():.2f} phút. Nên xem xét quan hệ phi tuyến (binning/spline/tree-based model) thay vì mặc định ảnh hưởng tuyến tính.")
# Nhận xét 4: thời tiết.
if arr_delay_col is not None and 'weather_corr' in globals() and not weather_corr.empty:
    w = weather_corr.iloc[0]
    findings.append(f"**Thông tin thời tiết:** biến `{w['Đặc trưng thời tiết']}` có |r| lớn nhất trong nhóm thời tiết (r = {w['Pearson r với ARR_DELAY']:.3f}). Cần đánh giá đóng góp bằng validation và tránh diễn giải tương quan thành nhân quả.")
# Nhận xét 5: nhãn mất cân bằng.
if label_col is not None:
    cc = analysis_df[label_col].dropna().astype(str).value_counts()
    if len(cc) >= 2:
        ratio = cc.max()/max(cc.min(), 1)
        findings.append(f"**Phân bố nhãn:** tỷ lệ lớp lớn nhất/lớp nhỏ nhất là {ratio:.2f}:1. Nếu mô hình phân loại, nên báo cáo precision/recall/F1 hoặc PR-AUC ngoài accuracy; cân nhắc stratified split và xử lý mất cân bằng chỉ trên tập train.")
# Nhận xét bổ sung về cách chia dữ liệu.
findings.append("**Chiến lược đánh giá:** dữ liệu chuyến bay có thể lặp lại theo ngày, hãng và tuyến; nên cân nhắc chia train/test theo thời gian (hoặc nhóm theo ngày/tuyến phù hợp mục tiêu) để giảm rò rỉ thông tin giữa các tập. Chỉ chuẩn hóa, mã hóa, chọn đặc trưng và cân bằng lớp dựa trên train.")

if not findings:
    findings = ["Chưa đủ cột để tự động tạo nhận xét định lượng. Kiểm tra cột mục tiêu và kiểu dữ liệu ở các mục phía trên.", "Ưu tiên chọn mục tiêu rõ ràng, xác minh các biến đầu vào có sẵn tại thời điểm dự báo, và chia tập kiểm thử độc lập.", "Sau khi nạp dữ liệu, chạy lại notebook để sinh các nhận xét dựa trên thống kê thực tế."]
for i, item in enumerate(findings[:6], start=1):
    display(Markdown(f"{i}. {item}"))
```

## 6. Phát hiện vấn đề chất lượng dữ liệu (tiêu chí 6 – 1,0 điểm)

Các chỉ số gồm giá trị thiếu, bản ghi trùng, ngoại lai theo quy tắc IQR, sai định dạng/giá trị ngoài miền hợp lệ, mất cân bằng nhãn và nguy cơ rò rỉ dữ liệu. Ngoại lai chỉ là **điểm cần xem xét**, không mặc định là lỗi: chậm chuyến rất lớn vẫn có thể là sự kiện có thật.

```python
quality_rows = []
# 6.1 Missing và dtype theo từng cột.
for c in df.columns:
    quality_rows.append({
        "Nhóm vấn đề": "Giá trị thiếu",
        "Cột/kiểm tra": c,
        "Số trường hợp": int(df[c].isna().sum()),
        "Tỷ lệ (%)": round(df[c].isna().mean()*100, 3),
        "Mức độ/ghi chú": "Cần đánh giá cơ chế thiếu; không tự động điền giá trị."
    })

# 6.2 Bản ghi trùng toàn bộ, không tính cột nguồn tạo bởi notebook.
quality_df = df.drop(columns=["_SOURCE_FILE"], errors="ignore")
full_dups = int(quality_df.duplicated().sum())
quality_rows.append({"Nhóm vấn đề": "Bản ghi trùng toàn bộ", "Cột/kiểm tra": "Toàn bộ cột dữ liệu", "Số trường hợp": full_dups, "Tỷ lệ (%)": round(full_dups/max(len(df),1)*100, 3), "Mức độ/ghi chú": "Kiểm tra trùng chính xác; đối chiếu nghiệp vụ trước khi loại bỏ."})

# 6.3 Trùng theo khóa ứng viên (không chắc chắn là khóa duy nhất).
key_names = ["MONTH", "DAY_OF_MONTH", "OP_UNIQUE_CARRIER", "ORIGIN", "DEST", "CRS_DEP_TIME"]
key_cols = [get_col(n) for n in key_names if get_col(n) is not None]
if len(key_cols) >= 3:
    key_dup_mask = df.duplicated(subset=key_cols, keep=False)
    key_dup_rows = int(key_dup_mask.sum())
    quality_rows.append({"Nhóm vấn đề": "Trùng khóa ứng viên", "Cột/kiểm tra": ", ".join(map(str,key_cols)), "Số trường hợp": key_dup_rows, "Tỷ lệ (%)": round(key_dup_rows/max(len(df),1)*100, 3), "Mức độ/ghi chú": "Chỉ là khóa ứng viên; cùng giờ/tuyến/hãng có thể có nhiều chuyến hợp lệ."})

# 6.4 Ngoại lai bằng IQR; bỏ mã thời gian/ngày tháng vì không phù hợp xem như biến liên tục.
noncontinuous = {get_col("MONTH"), get_col("DAY_OF_MONTH"), get_col("DAY_OF_WEEK"), get_col("CRS_DEP_TIME"), get_col("CRS_ARR_TIME"), "_SOURCE_FILE"}
outlier_summary = []
for c in df.columns:
    if c in noncontinuous:
        continue
    x = pd.to_numeric(df[c], errors="coerce")
    if x.notna().sum() < 20 or x.nunique(dropna=True) < 5:
        continue
    q1, q3 = x.quantile(.25), x.quantile(.75)
    iqr = q3 - q1
    if not np.isfinite(iqr) or iqr == 0:
        continue
    lo, hi = q1 - 1.5*iqr, q3 + 1.5*iqr
    mask = x.notna() & ((x < lo) | (x > hi))
    n = int(mask.sum())
    outlier_summary.append({"Cột/kiểm tra": c, "Số trường hợp": n, "Tỷ lệ (%)": round(n/max(x.notna().sum(),1)*100, 3), "Ngưỡng IQR thấp": lo, "Ngưỡng IQR cao": hi})
    quality_rows.append({"Nhóm vấn đề": "Ngoại lai (IQR)", "Cột/kiểm tra": c, "Số trường hợp": n, "Tỷ lệ (%)": round(n/max(x.notna().sum(),1)*100, 3), "Mức độ/ghi chú": f"Quy tắc 1.5×IQR: dưới {lo:.3f} hoặc trên {hi:.3f}; cần kiểm chứng nghiệp vụ."})
outlier_table = pd.DataFrame(outlier_summary)

# 6.5 Sai định dạng và ngoài miền hợp lệ cho các cột chuyến bay/thời tiết.
validation_specs = []
def add_invalid_check(col_name, condition_mask, description):
    if col_name is None:
        return
    n = int(condition_mask.fillna(False).sum()) if hasattr(condition_mask, "fillna") else int(np.asarray(condition_mask).sum())
    denom = max(int(df[col_name].notna().sum()), 1)
    validation_specs.append({"Cột/kiểm tra": col_name, "Số trường hợp": n, "Tỷ lệ (%)": round(100*n/denom, 3), "Mô tả lỗi/giới hạn": description})
    quality_rows.append({"Nhóm vấn đề": "Sai định dạng/ngoài miền", "Cột/kiểm tra": col_name, "Số trường hợp": n, "Tỷ lệ (%)": round(100*n/denom, 3), "Mức độ/ghi chú": description})

def numeric_col(nm):
    c = get_col(nm)
    if c is None: return None, None
    return c, pd.to_numeric(df[c], errors="coerce")

# Kiểm tra parse số: giá trị gốc không thiếu nhưng chuyển số bị NaN.
expected_numeric = ["MONTH", "DAY_OF_MONTH", "DAY_OF_WEEK", "CRS_DEP_TIME", "CRS_ARR_TIME", "ARR_DELAY", "CRS_ELAPSED_TIME", "DISTANCE", "ORIGIN_TEMP_C", "ORIGIN_PRECIPITATION_MM", "ORIGIN_WIND_SPEED_10M_KMH", "DEST_TEMP_C", "DEST_PRECIPITATION_MM", "DEST_WIND_SPEED_10M_KMH"]
for nm in expected_numeric:
    c = get_col(nm)
    if c:
        original_nonnull = df[c].notna()
        parsed = pd.to_numeric(df[c], errors="coerce")
        bad_parse = original_nonnull & parsed.isna()
        add_invalid_check(c, bad_parse, "Giá trị không thiếu nhưng không chuyển đổi được sang số; cần kiểm tra dấu thập phân/ký tự lạ.")

for nm, low, high, desc in [
    ("MONTH", 1, 12, "Tháng phải nằm trong [1, 12]."),
    ("DAY_OF_MONTH", 1, 31, "Ngày trong tháng phải nằm trong [1, 31]; tính hợp lệ còn phụ thuộc tháng."),
    ("DAY_OF_WEEK", 1, 7, "Thứ trong tuần phải nằm trong [1, 7]."),
    ("DISTANCE", 0, None, "Khoảng cách phải lớn hơn 0."),
    ("CRS_ELAPSED_TIME", 0, None, "Thời lượng bay dự kiến phải lớn hơn 0."),
    ("ORIGIN_PRECIPITATION_MM", 0, None, "Lượng mưa không được âm."),
    ("DEST_PRECIPITATION_MM", 0, None, "Lượng mưa không được âm."),
    ("ORIGIN_WIND_SPEED_10M_KMH", 0, None, "Tốc độ gió không được âm."),
    ("DEST_WIND_SPEED_10M_KMH", 0, None, "Tốc độ gió không được âm."),
]:
    c, x = numeric_col(nm)
    if c:
        if high is None:
            invalid = x <= low
        else:
            invalid = (x < low) | (x > high)
        add_invalid_check(c, invalid, desc)

# HHMM: giờ 0–23 và phút 0–59 (giá trị thiếu/bad parse đã được tính riêng).
for nm in ["CRS_DEP_TIME", "CRS_ARR_TIME"]:
    c, x = numeric_col(nm)
    if c:
        val = x.dropna()
        hh = np.floor(val / 100)
        mm = val % 100
        invalid_vals = val[(hh < 0) | (hh > 23) | (mm < 0) | (mm > 59) | (val % 1 != 0)]
        idxmask = df.index.isin(invalid_vals.index)
        add_invalid_check(c, pd.Series(idxmask, index=df.index), "Giờ dạng HHMM cần có giờ 0–23, phút 0–59 và không có phần thập phân.")

# Mã sân bay/hãng: phát hiện chuỗi rỗng hoặc mã không viết hoa/độ dài đáng ngờ, không tự sửa.
for nm, pattern, desc in [
    ("ORIGIN", r"^[A-Z]{3}$", "Mã sân bay đi thường có 3 chữ cái in hoa; kiểm tra giá trị lệch chuẩn."),
    ("DEST", r"^[A-Z]{3}$", "Mã sân bay đến thường có 3 chữ cái in hoa; kiểm tra giá trị lệch chuẩn."),
    ("OP_UNIQUE_CARRIER", r"^[A-Z0-9]{2,3}$", "Mã hãng bay thường dài 2–3 ký tự in hoa/số; kiểm tra giá trị lệch chuẩn.")
]:
    c = get_col(nm)
    if c:
        s = df[c].astype("string")
        invalid = s.notna() & ~s.str.strip().str.match(pattern, na=False)
        add_invalid_check(c, invalid, desc)

quality_table = pd.DataFrame(quality_rows)
# Bảng tổng hợp các vấn đề có số lượng > 0; bảng missing theo từng cột được giữ riêng để không quá dài.
issue_overview = quality_table[quality_table["Nhóm vấn đề"].ne("Giá trị thiếu")].copy()
display(Markdown("### Bảng thống kê vấn đề chất lượng dữ liệu"))
display(issue_overview.sort_values(["Nhóm vấn đề", "Số trường hợp"], ascending=[True, False]).reset_index(drop=True))
display(Markdown("### Chi tiết giá trị thiếu theo cột"))
missing_table = quality_table[quality_table["Nhóm vấn đề"].eq("Giá trị thiếu")][["Cột/kiểm tra", "Số trường hợp", "Tỷ lệ (%)", "Mức độ/ghi chú"]].sort_values("Tỷ lệ (%)", ascending=False)
display(missing_table)
if not outlier_table.empty:
    display(Markdown("### Chi tiết ngoại lai theo IQR"))
    display(outlier_table.sort_values("Tỷ lệ (%)", ascending=False).reset_index(drop=True))
if validation_specs:
    display(Markdown("### Chi tiết sai định dạng và giá trị ngoài miền"))
    display(pd.DataFrame(validation_specs).sort_values("Số trường hợp", ascending=False).reset_index(drop=True))

# Lưu bảng thống kê thành CSV để dùng kèm báo cáo nếu cần.
REPORT_DIR.mkdir(parents=True, exist_ok=True)
issue_overview.to_csv(REPORT_DIR / "bang_van_de_chat_luong.csv", index=False, encoding="utf-8-sig")
missing_table.to_csv(REPORT_DIR / "bang_gia_tri_thieu.csv", index=False, encoding="utf-8-sig")
print("Đã lưu bảng chất lượng vào thư mục reports/.")
```

### 6.6. Mất cân bằng lớp và nguy cơ rò rỉ dữ liệu

- **Mất cân bằng lớp:** định lượng bằng số lượng và tỷ lệ mỗi lớp. Nếu nhãn phân loại được tạo từ `ARR_DELAY`, ngưỡng phân loại được ghi rõ là trên 15 phút.
- **Rò rỉ dữ liệu:** xác định các cột có tên/ý nghĩa thường chỉ biết sau chuyến bay hoặc được tính trực tiếp từ nhãn. Đây là bước rà soát ban đầu; cần căn cứ vào thời điểm dự báo thực tế để quyết định loại cột.
- Nếu mô hình dự đoán `ARR_DELAY`, không đưa chính `ARR_DELAY` hay nhãn phân loại dẫn xuất từ nó vào tập đặc trưng X. Những biến quan sát sau khi chuyến bay diễn ra có thể gây rò rỉ mục tiêu.

```python
# Mất cân bằng lớp: dùng nhãn có sẵn hoặc nhãn phân tích đã tạo ở mục 4.
if label_col is not None:
    class_counts = analysis_df[label_col].dropna().astype(str).value_counts().rename_axis("Nhãn").reset_index(name="Số bản ghi")
    class_counts["Tỷ lệ (%)"] = (class_counts["Số bản ghi"] / class_counts["Số bản ghi"].sum() * 100).round(3)
    display(class_counts)
    if len(class_counts) >= 2:
        ratio = class_counts["Số bản ghi"].max() / max(class_counts["Số bản ghi"].min(), 1)
        majority_share = class_counts["Tỷ lệ (%)"].max()
        if ratio >= 3 or majority_share >= 75:
            imbalance_note = f"Có dấu hiệu mất cân bằng lớp: lớp nhiều nhất chiếm {majority_share:.2f}% và tỷ lệ lớp lớn nhất/nhỏ nhất là {ratio:.2f}:1. Với bài toán phân loại, nên dùng stratified split, báo cáo macro-F1/recall/PR-AUC và thử class weights; mọi biện pháp cân bằng chỉ áp dụng trên tập train."
        else:
            imbalance_note = f"Theo ngưỡng rà soát thô (tỷ lệ lớp lớn nhất/nhỏ nhất < 3 và lớp lớn nhất < 75%), chưa thấy mất cân bằng nghiêm trọng: tỷ lệ là {ratio:.2f}:1. Vẫn cần xem phân bố riêng trong train/test."
        comment(imbalance_note)
    else:
        comment("Nhãn chỉ có một giá trị quan sát được; không thể huấn luyện bộ phân loại hữu ích cho đến khi kiểm tra lại cách tạo nhãn hoặc dữ liệu.")
else:
    display(Markdown("Không thể định lượng mất cân bằng lớp vì không tìm thấy cột nhãn và không có ARR_DELAY để tạo nhãn phân tích."))

# Rà soát nguy cơ rò rỉ theo tên cột và thời điểm sẵn có.
leakage_patterns = {
    "Rất cao – có thể là nhãn/biến dẫn xuất từ nhãn": [r"^ARR_DELAY$", r"ARR_DELAY_NEW", r"ARR_DEL15", r"ARR_DELAY_GROUP", r"DELAYED_?15M", r"IS_DELAYED", r"DELAY_LABEL", r"^TARGET$", r"^LABEL$"],
    "Cao – có thể chỉ biết sau khi chuyến bay diễn ra": [r"^ARR_TIME$", r"ACTUAL_ARR", r"ACTUAL_ELAPSED", r"WHEELS_ON", r"WHEELS_OFF", r"TAXI_IN", r"TAXI_OUT", r"^DEP_DELAY$", r"^DEP_TIME$", r"CANCELL", r"DIVERT", r"AIR_TIME"]
}
leak_rows = []
for c in df.columns:
    if c == "_SOURCE_FILE":
        continue
    upper = str(c).upper()
    level = None
    reason = None
    for lvl, patterns in leakage_patterns.items():
        if any(re.search(p, upper) for p in patterns):
            level = lvl
            reason = "Tên cột khớp quy tắc rà soát; xác minh biến có sẵn tại thời điểm dự đoán."
            break
    if level:
        if arr_delay_col is not None and c == arr_delay_col:
            reason = "Đây là biến mục tiêu ví dụ; tuyệt đối không đưa vào X khi dự đoán ARR_DELAY, nhưng giữ lại để làm y."
        leak_rows.append({"Cột": c, "Mức rủi ro sơ bộ": level, "Lý do/việc cần làm": reason})

if leak_rows:
    leakage_table = pd.DataFrame(leak_rows)
    display(leakage_table)
else:
    leakage_table = pd.DataFrame(columns=["Cột", "Mức rủi ro sơ bộ", "Lý do/việc cần làm"])
    display(Markdown("Không phát hiện cột khớp các quy tắc tên thường gặp. **Điều này không chứng minh dữ liệu hoàn toàn không rò rỉ**; cần xác minh theo quy trình tạo dữ liệu và thời điểm dự báo."))

display(Markdown("**Nguy cơ theo thiết kế đánh giá:** nếu các chuyến bay cùng ngày/tuyến/hãng có đặc điểm gần nhau, chia ngẫu nhiên từng dòng có thể khiến train và test quá giống nhau. Nên cân nhắc kiểm thử theo thời gian hoặc theo nhóm phù hợp ứng dụng thực tế."))
```

## 7. Đối chiếu dữ liệu thô và dữ liệu đã xử lý

Phần này cho biết thay đổi về số dòng/cột và các vấn đề chất lượng trước/sau xử lý. Nếu hai thư mục chứa các phạm vi thời gian khác nhau, chênh lệch số dòng không thể tự động được xem là số bản ghi đã loại bỏ; cần kiểm tra tên tệp và quy trình làm sạch.

```python
comparison_rows = []
for label, frame, files in [("Raw", raw_df, raw_files), ("Đã xử lý", processed_df, processed_files)]:
    if frame.empty:
        comparison_rows.append({"Bộ dữ liệu": label, "Số tệp": len(files), "Số dòng": 0, "Số cột (gồm nguồn)": 0, "Số ô thiếu": 0, "Tỷ lệ ô thiếu (%)": np.nan, "Bản ghi trùng toàn bộ": 0})
    else:
        x = frame.drop(columns=["_SOURCE_FILE"], errors="ignore")
        cells_count = x.shape[0] * x.shape[1]
        comparison_rows.append({
            "Bộ dữ liệu": label,
            "Số tệp": len(files),
            "Số dòng": len(frame),
            "Số cột (gồm nguồn)": x.shape[1],
            "Số ô thiếu": int(x.isna().sum().sum()),
            "Tỷ lệ ô thiếu (%)": round(100*x.isna().sum().sum()/cells_count, 3) if cells_count else 0,
            "Bản ghi trùng toàn bộ": int(x.duplicated().sum())
        })
comparison = pd.DataFrame(comparison_rows)
display(comparison)

if not raw_df.empty and not processed_df.empty:
    raw_cols = set(raw_df.columns) - {"_SOURCE_FILE"}
    proc_cols = set(processed_df.columns) - {"_SOURCE_FILE"}
    raw_n, proc_n = len(raw_df), len(processed_df)
    row_delta = proc_n - raw_n
    col_added = sorted(proc_cols - raw_cols)
    col_removed = sorted(raw_cols - proc_cols)
    display(Markdown(f"- Chênh lệch số dòng (đã xử lý − raw): **{row_delta:+,}**. Đây có thể do lọc bản ghi, khác phạm vi tệp hoặc lỗi nạp; cần đối chiếu quy trình xử lý.\n- Cột chỉ có trong dữ liệu đã xử lý: **{', '.join(col_added) if col_added else 'Không có'}**.\n- Cột chỉ có trong dữ liệu raw: **{', '.join(col_removed) if col_removed else 'Không có'}**.\n- Không nên kết luận dữ liệu đã tốt hơn chỉ từ số dòng; cần so sánh các chỉ số thiếu, trùng, định dạng và kiểm tra tính hợp lệ ở trên."))
else:
    display(Markdown("Cần có dữ liệu ở cả hai thư mục để tạo đối chiếu raw – đã xử lý đầy đủ."))
```

## 8. Kết luận EDA và khuyến nghị cho xây dựng mô hình

Phần kết luận được sinh theo số liệu đã đọc, để có thể sử dụng làm bản nháp mục nhận xét trong báo cáo. Hãy đọc lại trước khi nộp và bổ sung bối cảnh nghiệp vụ/định nghĩa nhãn mà nhóm sử dụng.

```python
# Sinh kết luận ngắn gọn, bám theo thống kê thực tế.
conclusion_lines = []
conclusion_lines.append(f"**Quy mô:** dữ liệu dùng cho EDA gồm {len(df):,} dòng và {len([c for c in df.columns if c != '_SOURCE_FILE'])} cột dữ liệu.")
missing_total = int(df.drop(columns=["_SOURCE_FILE"], errors="ignore").isna().sum().sum())
total_cells = int(len(df) * len([c for c in df.columns if c != '_SOURCE_FILE']))
missing_rate = 100*missing_total/max(total_cells, 1)
conclusion_lines.append(f"**Thiếu dữ liệu:** tổng tỷ lệ ô thiếu là {missing_rate:.2f}%; cần quyết định giữ/loại/điền theo từng cột và chỉ fit phép điền trên tập train.")
conclusion_lines.append(f"**Trùng lặp:** có {full_dups:,} bản ghi trùng toàn bộ theo các cột dữ liệu đang nạp (không tính `_SOURCE_FILE`). Kiểm tra thêm khóa nghiệp vụ trước khi loại bỏ.")
if arr_delay_col is not None:
    delay_values = pd.to_numeric(df[arr_delay_col], errors="coerce").dropna()
    if not delay_values.empty:
        delay15 = int((delay_values > DELAY_THRESHOLD_MINUTES).sum())
        conclusion_lines.append(f"**Đầu ra ARR_DELAY:** trung bình {delay_values.mean():.2f} phút, trung vị {delay_values.median():.2f} phút; {delay15:,}/{len(delay_values):,} chuyến ({100*delay15/len(delay_values):.2f}%) trễ hơn {DELAY_THRESHOLD_MINUTES} phút theo nhãn phân tích.")
if label_col is not None:
    value_counts = analysis_df[label_col].dropna().astype(str).value_counts()
    if len(value_counts) >= 2:
        conclusion_lines.append(f"**Cân bằng lớp:** tỷ lệ lớp lớn nhất/nhỏ nhất là {value_counts.max()/max(value_counts.min(),1):.2f}:1; cân nhắc đánh giá bằng metric phù hợp nếu xây dựng bài toán phân loại.")
if not outlier_table.empty:
    top_out = outlier_table.sort_values("Tỷ lệ (%)", ascending=False).iloc[0]
    conclusion_lines.append(f"**Ngoại lai IQR:** cột `{top_out['Cột/kiểm tra']}` có tỷ lệ điểm ngoài 1.5×IQR cao nhất trong các biến đã kiểm tra ({top_out['Tỷ lệ (%)']:.2f}%). Không nên xóa tự động vì các giá trị cực đoan có thể có ý nghĩa nghiệp vụ.")
if not leakage_table.empty:
    conclusion_lines.append(f"**Rò rỉ dữ liệu:** có {len(leakage_table)} cột cần rà soát theo quy tắc tên; loại mục tiêu khỏi X và chỉ dùng biến thực sự sẵn có tại thời điểm dự báo.")
else:
    conclusion_lines.append("**Rò rỉ dữ liệu:** chưa phát hiện cột theo các quy tắc tên thông dụng; vẫn phải xác minh thời điểm phát sinh dữ liệu và tránh chia train/test gây trùng lặp thông tin.")
conclusion_lines.extend([
    "**Khuyến nghị tiền xử lý:** xử lý sai định dạng trước; quyết định giá trị thiếu theo từng trường; kiểm tra trùng theo khóa nghiệp vụ; giữ và phân tích riêng ngoại lai trước khi cân nhắc biến đổi/winsorize.",
    "**Khuyến nghị mô hình:** nếu dự báo ARR_DELAY, coi đây là hồi quy và đánh giá MAE/RMSE cùng baseline. Nếu chuyển thành phân loại trễ trên 15 phút, ghi rõ định nghĩa nhãn, dùng stratified split và báo cáo precision/recall/F1/PR-AUC.",
    "**Khuyến nghị đánh giá:** ưu tiên chia tập theo thời gian nếu mục tiêu là dự đoán chuyến bay tương lai; thực hiện mã hóa, chuẩn hóa, chọn đặc trưng và xử lý mất cân bằng chỉ trong pipeline fit trên tập train."
])
display(Markdown("\n\n".join(f"- {line}" for line in conclusion_lines)))

# Xuất các bảng chính để đính kèm báo cáo.
if not target_corr.empty if 'target_corr' in globals() else False:
    target_corr.rename("Pearson_r_ARR_DELAY").to_csv(REPORT_DIR / "tuong_quan_voi_arr_delay.csv", encoding="utf-8-sig")
if not comparison.empty:
    comparison.to_csv(REPORT_DIR / "doi_chieu_raw_processed.csv", index=False, encoding="utf-8-sig")
print(f"Hoàn tất. Hình ở: {FIGURE_DIR}")
print(f"Bảng CSV báo cáo ở: {REPORT_DIR}")
```
